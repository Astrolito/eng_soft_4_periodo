# 🔄 Controle de Versão com GIT

**O que é Controle de Versão?**
Sistema que registra mudanças em arquivos ao longo do tempo, permitindo recuperar versões anteriores, trabalhar em equipe e rastrear quem fez o quê e quando.

## 📊 Diferenças entre Controle de Versão

### Sistemas Centralizados (SVN, CVS, Perforce)
* Um único servidor contém todos os arquivos de controle de versão
* Vários desenvolvedores usam arquivos a partir desse lugar central
* Se o disco for corrompido, é muito difícil recuperar o conteúdo
* Requer acesso ao repositório remoto (internet necessária)
* Foi o padrão por muito tempo

### Git - Sistema Distribuído ✓
* Desenvolvido em 2005 por Linus Torvalds (criador do Linux)
* Código aberto (Open Source)
* **Cada desenvolvedor tem uma cópia completa do projeto** (backup automático)
* Cada clone é um backup dos dados (maior segurança)
* Funciona offline (não necessita internet para trabalhar localmente)
* Permite múltiplos fluxos de trabalho e metodologias diferentes
* Eficiente em projetos pequenos e grandes
* Compatível com muitos sistemas e protocolos

### Características de Segurança do Git
* Todo conteúdo (arquivos, versões, commits, diretórios) é protegido com **SHA1** (algoritmo de hash criptográfico)
* Impossível modificar o histórico sem detectar alterações
* Garante integridade dos dados

### Vantagens Principais do Git
* **Capacidade de criar pontos na história** (commits) - snapshots do projeto
* **Ramificações de trabalho** (branches) - trabalhe em features sem afetar código principal
* **Análise e resolução de conflitos** - quando duas pessoas modificam o mesmo arquivo
* **Desenvolvimento paralelo** - dois ou mais desenvolvedores no mesmo projeto simultaneamente
* **Trabalho offline** - crie commits sem conexão com servidor
* Centralizado em repositório como GitHub, GitLab, Bitbucket

## 📚 Conceitos Principais do Git

### 📁 **Repositório (Repository)**
* Local (físico ou remoto) onde os arquivos e histórico de versões são armazenados
* **Repositório Local**: Pasta .git no seu computador com todo histórico
* **Repositório Remoto**: Servidor central (GitHub, GitLab, etc.) compartilhado pela equipe
* Comando: `git init` (criar local) ou `git clone` (copiar remoto)

### 💾 **Commit (Confirmação)**
* Snapshot de uma versão específica do projeto
* Registra QUAL arquivo foi modificado, QUANDO e QUEM fez
* Inclui mensagem descritiva do que foi alterado
* Exemplo: "Commit: Corrigido bug no login"
* Comando: `git commit -m "descrição das mudanças"`

### 🌳 **Branch (Ramificação)**
* Cópia do código original que pode ser modificada **sem afetar a versão principal**
* Permite trabalho paralelo em diferentes features
* Após concluir a feature, faz-se merge para integrar na main
* **main/master**: Branch principal (produção)
* **develop**: Branch de desenvolvimento
* Exemplo: `feature/login`, `bugfix/database`, etc.
* Comando: `git branch nome-branch`

### 🔗 **Merge (Fusão)**
* Une as mudanças de uma branch com outra (geralmente na main)
* Combina o código e resolve conflitos se existirem
* Acontece primeiramente no repositório local
* Exemplo: Terminou feature no seu computador → merge para main localmente
* Comando: `git merge nome-branch`

### ⬆️ **Push (Envio)**
* Envia (empurra) seus commits locais para o repositório remoto
* Permite que outros desenvolvedores vejam suas mudanças
* Pré-requisito: Ter feito merge na branch remota
* Exemplo: Desenvolveu localmente → push para GitHub
* Comando: `git push origin nome-branch`

### ⬇️ **Pull (Recepção)**
* Baixa (puxa) as modificações do repositório remoto para sua máquina
* Atualiza seu código local com mudanças de outros desenvolvedores
* Combina fetch (baixar) + merge (integrar) em um comando
* Exemplo: Colega fez push → você faz pull para ter as mudanças
* Comando: `git pull origin nome-branch`

## 🔄 Fluxo Básico de Trabalho com Git

```
1. git clone <repositório>         # Copiar projeto remoto
2. git branch <feature>             # Criar nova branch
3. git checkout <feature>           # Mudar para a branch
   (Editar arquivos)
4. git add .                        # Preparar mudanças
5. git commit -m "mensagem"        # Registrar mudanças
6. git push origin <feature>       # Enviar para remoto
7. git checkout main               # Voltar para main
8. git pull origin main            # Atualizar main local
9. git merge <feature>             # Integrar feature em main
10. git push origin main           # Enviar main atualizada
```

## 📋 Exemplo Prático Passo a Passo

### Cenário: Você quer adicionar uma nova funcionalidade de login

**Passo 1: Preparar o ambiente**
```bash
# Clonar repositório (primeira vez)
git clone https://github.com/usuario/projeto.git
cd projeto

# Ou atualizar código local (próximas vezes)
git pull origin main
```

**Passo 2: Criar branch para sua feature**
```bash
# Criar e entrar em uma branch de feature
git checkout -b feature/autenticacao
# Equivalente a:
# git branch feature/autenticacao
# git checkout feature/autenticacao
```

**Passo 3: Desenvolver (local)**
```
Você cria/edita arquivos:
- login.py
- auth.html
- database.sql
```

**Passo 4: Registrar mudanças**
```bash
# Ver quais arquivos foram modificados
git status

# Preparar mudanças para commit
git add .

# Registrar mudanças com mensagem descritiva
git commit -m "Adiciona sistema de autenticação com login"
```

**Passo 5: Enviar para repositório remoto**
```bash
# Fazer push da branch para GitHub
git push origin feature/autenticacao
```

**Passo 6: Integrar na versão principal**
```bash
# Voltar para a branch main
git checkout main

# Atualizar main local com mudanças remotas
git pull origin main

# Integrar sua feature na main
git merge feature/autenticacao

# Enviar a main atualizada
git push origin main
```

### Status do Repositório Antes e Depois

```
ANTES (sem branch):
main → login_v1 → login_v2 → login_v3 (todos modificados por você)

DEPOIS (com branch):
main ─────────────────────────── (original, seguro)
       └─ feature/autenticacao ─────── (seu trabalho isolado)
                                 └─────→ (merge de volta à main)
```

## ⚠️ Situações Comuns e Como Resolvê-las

### Conflito de Merge
**O que é?** Dois desenvolvedores modificaram a mesma linha do mesmo arquivo.

**Como resolver:**
```bash
# Git marca as linhas em conflito
<<<<<<< HEAD
    seu código aqui
=======
    código do colega aqui
>>>>>>> feature/outra-pessoa

# Você escolhe qual manter, edita e faz:
git add .
git commit -m "Resolvido conflito em arquivo.py"
```

### Você fez push mas precisa desfazer
```bash
# Ver histórico de commits
git log

# Voltar a um commit anterior
git revert <id-do-commit>
```

### Seus arquivos foram modificados mas quer descartar mudanças
```bash
# Descartar todas as mudanças locais
git checkout .

# Descartar mudanças de um arquivo específico
git checkout -- nome-arquivo.py
```

## ✅ Boas Práticas com Git

### 1. **Commits Frequentes e Descritivos**
- ❌ RUIM: `git commit -m "atualizações"`
- ✅ BOM: `git commit -m "Corrigido bug de validação no formulário de login"`

### 2. **Use Branches para Cada Feature**
- Nunca trabalhe direto em `main` ou `master`
- Use nomes descritivos: `feature/pagamento`, `bugfix/validacao`, `hotfix/segurança`

### 3. **Sempre Faça Pull Antes de Push**
```bash
# Antes de fazer push, sempre faça pull para ter mudanças dos colegas
git pull origin main
git push origin feature/sua-feature
```

### 4. **Nunca Force Push sem Necessidade**
- `git push -f` sobrescreve histórico remoto
- Pode apagar trabalho de outras pessoas!
- Use apenas em emergências e com comunicação com a equipe

### 5. **Escreva Mensagens de Commit Claras**
```
Estrutura recomendada:
- Primeira linha: Resumo (máx 50 caracteres)
- Linha vazia
- Corpo: Detalhes da mudança

Exemplo:
Adiciona autenticação com JWT

- Implementa sistema de tokens JWT
- Valida tokens em cada requisição
- Refresh tokens com expiração de 24h
- Testes unitários inclusos
```

### 6. **Revise seu código antes de Commit**
```bash
# Ver mudanças antes de fazer commit
git diff

# Ver mudanças apenas de um arquivo
git diff nome-arquivo.py
```

## Exercícios Par ou Ímpar

**Conceito**: Um número é **par** se é divisível por 2 (resto 0) e **ímpar** se deixa resto 1.

**Operador Módulo (%)**: Retorna o resto da divisão. Exemplo:
- `10 % 2 = 0` (10 é par)
- `7 % 2 = 1` (7 é ímpar)

````python
numero = int(input("Insira um número: "))

# Verifica o resto da divisão por 2
if numero % 2 == 0:
    print(f"O número {numero} é par.")
else:
    print(f"O número {numero} é ímpar.")
````

## Exercício: Média de 3 Notas

**Conceitos importantes**:
- **float**: Números com casas decimais (ex: 7.5, 8.2)
- **Divisão normal (/)**: Sempre retorna número com decimal
- **Divisão inteira (//)**: Arredonda para baixo (perde casas decimais)

**Nota importante**: Para calcular média corretamente, use `/` e não `//`

````python
N1 = float(input("Insira a primeira nota: "))
N2 = float(input("Insira a segunda nota: "))
N3 = float(input("Insira a terceira nota: "))

# Cálculo da média (usando / para obter resultado decimal)
media = (N1 + N2 + N3) / 3
print(f"A média dos números é: {media:.2f}")  # .2f exibe 2 casas decimais

# Verifica se aluno foi aprovado ou reprovado
if media >= 7:
    print(f"Aluno com média {media:.2f}: Aprovado")
else: 
    print(f"Aluno com média {media:.2f}: Reprovado")
````

# Estruturas de Repetição (Loops)

## Loop FOR

O laço **FOR** é a estrutura de repetição mais utilizada em Python. Executa um bloco de código um número determinado de vezes.

**Sintaxe**: 
```python
for <variável> in range(inicio, limite, passo):
    # código a executar
```

**A função range()**: Cria uma sequência de números
- `range(n)` → de 0 a n-1
- `range(i, n)` → de i a n-1
- `range(i, n, p)` → de i a n-1, contando de p em p

**Exemplos**:
- `range(5)` → 0, 1, 2, 3, 4
- `range(2, 8)` → 2, 3, 4, 5, 6, 7
- `range(0, 10, 2)` → 0, 2, 4, 6, 8

````python
# Exemplo 1: Contagem de 1 a 10
print("Contando de 1 a 10:")
for i in range(1, 11):  # de 1 até 10 (11 é excluído)
    print(i, end=" ")  # end=" " coloca espaço em vez de quebra de linha

print("\n")  # quebra de linha

# Exemplo 2: Contagem de 2 em 2
print("Contando de 0 a 10, de 2 em 2:")
for i in range(0, 11, 2):  # o terceiro parâmetro é o passo (de quanto em quanto)
    print(i, end=" ")
````

````python
# Exemplo 3: Iterando sobre uma lista
print("\nIterando sobre uma lista:")
vet = [1, 2, 3, 4, 5]
for i in vet:
    print(f"Elemento: {i}")

# Exemplo 4: Usando índices
print("\nUsando índices (posição dos elementos):")
for i in range(len(vet)):
    print(f"Posição {i}: {vet[i]}")
````

## Instruções Break e Continue

Permitem melhor controle sobre loops:

- **`break`**: Interrompe o loop **completamente** (sai dele)
- **`continue`**: Pula a iteração atual e vai para a **próxima** (não termina o loop)

**Quando usar**:
- `break`: Encontrou o que procurava ou ocorreu erro
- `continue`: Quer ignorar certos valores e continuar processando outros

````python
# Exemplo de break
print("Contando até encontrar 3 (depois para):")
for x in range(0, 11):
    if x == 3:
        print(f"Encontrou {x}! Saindo do loop...")
        break  # sai do loop completamente
    print(x, end=" ")

print("\n")
````

````python
# Exemplo de continue
print("Contando de 0 a 10, ignorando 3 e 7:")
for x in range(0, 11):
    if x == 3 or x == 7:
        continue  # pula esta iteração, vai para a próxima
    print(x, end=" ")

print("\n")
````

## Loop WHILE

O comando **while** executa um bloco de código **enquanto uma condição for verdadeira**.

**Sintaxe**: 
```python
while <condição>:
    # bloco de código
    # IMPORTANTE: deve alterar algo para evitar loop infinito!
```

**Diferença FOR vs WHILE**:
- **FOR**: Usamos quando sabemos **quantas** vezes repetir
- **WHILE**: Usamos quando repetimos **até uma condição ser falsa**

**Atenção**: Se a condição nunca virar falsa, cria um **loop infinito** (programa trava)!

````python
# Exemplo 1: Contagem simples com while
print("Contando de 0 a 9:")
inicio = 0
while inicio < 10:  # enquanto inicio for menor que 10
    print(inicio, end=" ")
    inicio += 1  # IMPORTANTE: incrementar para não criar loop infinito

print("\n")
````

````python
# Exemplo 2: Validação de senha
print("Exemplo: Verificação de Senha")
senha_correta = "54321"
leitura = input("Digite a senha: ")
while leitura != senha_correta:
    print("❌ Senha incorreta!")
    leitura = input("Digite a senha novamente: ")
print("✓ Senha correta! Acesso concedido.")
````

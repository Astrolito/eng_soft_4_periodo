## Exercicios Par ou impar

```python
numero = int(input("Insira um número: "))

if numero % 2 == 0:
    print(f"O número {numero} é par.")
else:
    print(f"O número {numero} é ímpar.")
```

## Exercicio: media de 3 notas


```python
N1 = float(input("Insira o primeiro número: "))
N2 = float(input("Insira o segundo número: "))
N3 = float(input("Insira o terceiro número: "))

media = (N1 + N2 + N3) // 3
print(f"A média dos números é: {media}")

if media >= 7:
    print(f"aluno com média {media}: Aprovado")
else: 
    print(f"aluno com média {media}: Reprovado")
```

# Estruturas de repeticao

laco FOR: e a estrutura de repeticao mais utilizada em python
Sintaxe: for <variavel> in range (inicio, limite, passo) ex: (0 #comeca em 0 , 9 #finaliza em 8 , 2 #conta de 2 em 2)
            print(variavel)

A função range() cria uma lista com os valores de um intervalo apontado por seus parâmetros.
• A função range() pode ter de 1 a 3 parâmetros:
• range(n) gera um intervalo de 0 a n-1
• range(i, n) gera um intervalo de i a n-1 range(i, n, p) gera um intervalo de i a n-1 com intervalo p entre os números

```python
#contagem
for i in range(1, 11, 1):#o ultimo numero é o passo, ou seja, de quanto em quanto ele vai contar
    print(i)
```

```python
vet = [1, 2, 3, 4, 5]
for i in vet:
    print(f"\n{i}")
```

Break e continue: • As repetições podem ser melhor utilizadas através das instruções break e continue. A instrução break interrompe o laço (terminando-o por completo) e a instrução continue pula para a próxima iteração imediatamente (não termina o laço, apenas passa à próxima iteração).

```python
#Exemplo de break

for x in range(0, 11):
    if x == 3:
        break
    print(x)
```

```python
#Exemplo de continue

for x in range(0, 11):
    if x == 3:
        continue
    print(x)
```

Em Python, o comando while é uma estrutura de repetição que executa um bloco de código enquanto uma condição for verdadeira.

Sintaxe básica

while <condição>:
    # bloco de código a ser repetido
    # (deve alterar algo para evitar loop infinito)

```python
inicio = 0
while inicio < 10:
    print(inicio)
    inicio += 1
```

```python
senha = "54321"
leitura = input("Digite a senha: ")
while leitura != senha:
    print("Senha incorreta!")
    leitura = input("Digite a senha: ")
print("Senha correta!")
```

```python
contador = 0
somador = 0
while contador < 5:
    numero = int(input("Digite um número: "))
    somador += numero
    contador += 1
print("A soma dos números é:", somador)
print("A média dos números é:", somador / contador)
```

Exercício 1: Cálculo de Bônus de Vendas (RH/Vendas): uma empresa decidiu dar
um bônus de 10% sobre o faturamento total para a equipe de vendas. Calcule o
valor do bônus e o faturamento final da empresa após subtrair esse bônus.
• Faturamento inicial: 50.000
• Percentual de bônus: 0.10

```python
#calculo de bonus

faturamento = 50000
bonus = 0.1

faturamento_bonus = faturamento * bonus
valor_bonus = faturamento + faturamento_bonus

print(f"O faturamento com bonus é: {faturamento_bonus}")
print(f"O valor total com bonus é: {valor_bonus}")
```

# Estrutura de Dados

As estruturas de dados definem a organização, a forma de acesso e os métodos de processamento disponíveis para um conjunto de elementos. Em Ciência da Computação, esse conceito está diretamente relacionado ao estudo de tipos abstratos de dados (TADs), cuja finalidade é representar informações de maneira eficiente, preservando propriedades semânticas e operacionais do problema.

## Conceito acadêmico
Uma estrutura de dados não é apenas um armazenamento de valores; ela estabelece:
* a ordem dos elementos;
* a forma de acesso aos elementos;
* a possibilidade de modificação;
* a complexidade de operações como busca, inserção e remoção;
* o uso de memória e a relação com algoritmos que manipulam esses dados.

A escolha apropriada da estrutura influencia diretamente a eficiência computacional de um sistema, pois diferentes estruturas apresentam distintos custos temporais e espaciais.

## Exemplos estudados
* Listas;
* Tuplas;
* Dicionários;
* Conjuntos;
* Entre outras estruturas que serão aprofundadas em cursos subsequentes.

## Importância em algoritmos e sistemas
A organização dos dados determina a facilidade de implementação e o desempenho dos algoritmos. Por exemplo, algoritmos de ordenação, busca e filtragem realizam melhoramentos significativos quando a estrutura escolhida favorece a operação necessária. Assim, o estudo das estruturas de dados é essencial para a análise de complexidade, otimização e projeto de software.


### Listas
Uma lista em Python é uma sequência ordenada e mutável de elementos. Em termos conceituais, trata-se de uma coleção linear indexada, em que cada item ocupa uma posição determinada e pode ser acessado por meio de um índice inteiro. O primeiro elemento possui índice 0 e o acesso é realizado em tempo constante em grande parte dos casos, embora operações como inserção ou remoção no meio da lista possam demandar deslocamento de elementos.

Em Python, a lista é representada com colchetes: nome_lista = [valor1, valor2, ..., valorN]. Essa estrutura permite armazenar elementos heterogêneos, isto é, valores de tipos distintos em uma mesma coleção.

* A lista é mutável, isto é, seus elementos podem ser alterados após a criação.
* Ela preserva a ordem de inserção dos elementos.
* Pode conter elementos repetidos.
* Pode armazenar valores de tipos diferentes, inclusive outras listas.
* É possível criar uma lista vazia com list() ou []

#### Operações fundamentais
A manipulação de listas é uma prática central em programação, pois ela permite a construção de algoritmos de processamento sequencial, filtragem, agregação e transformação de dados.

* Acesso por índice: lista[0]
* Fatiamento: lista[1:3]
* Adição de elemento: append()
* Inserção em posição específica: insert(posição, valor)
* Remoção por valor: remove(valor)
* Remoção por posição: pop()
* Ordenação: sort()
* Determinação do tamanho: len(lista)
* Contagem de ocorrências: lista.count(valor)
* Inversão da ordem: lista.reverse()

#### Comandos em Python
```python
lista = [10, 20, 30]
print(lista[0])          # acesso ao primeiro elemento
print(lista[1:3])        # fatiamento
lista.append(40)          # adiciona ao final
lista.insert(1, 15)       # insere na posição 1
lista.remove(20)          # remove o valor 20
lista.pop()               # remove o último elemento
lista.sort()              # ordena em ordem crescente
lista.reverse()           # inverte a ordem
print(lista)
print(len(lista))        # tamanho da lista
print(lista.count(15))   # quantas vezes o valor aparece
```

#### Exemplo ilustrativo
```python
alunos = ["Ana", "Bruno", "Carla"]
alunos.append("Davi")
alunos[0] = "Alice"
print(alunos)
print(alunos[1:3])
print(len(alunos))
print(sorted(alunos))
```

#### Análise conceitual
As listas são particularmente adequadas quando se exige atualização dinâmica de dados, como em cadastros, filas, registros em memória e manipulação de elementos em tempo de execução. A sua natureza mutável torna-as extremamente flexíveis, porém exige cuidado em relação à consistência dos dados e ao custo de operações de inserção e remoção em posições intermediárias.

#### Vantagens e uso prático
* Organização sequencial de dados;
* Facilidade de modificação e expansão;
* Aplicabilidade em estruturas dinâmicas e algoritmos de processamento;
* Apropriadas para dados cuja ordem de entrada é relevante.

> Importante: embora as listas sejam altamente versáteis, sua mutabilidade exige atenção na análise de complexidade e na manutenção de invariantes do programa.

## Tuplas
Uma tupla é uma estrutura linear imutável, isto é, após sua criação, seus elementos não podem ser alterados, adicionados ou removidos. Em termos formais, ela representa uma sequência ordenada de valores com restrição de mutabilidade. Em Python, a tupla é definida por meio de parênteses: nome_tupla = (valor1, valor2, valor3).

Essa característica torna a tupla uma alternativa apropriada quando o conjunto de dados deve permanecer consistente durante a execução do programa. Tuplas são amplamente utilizadas em contextos em que a integridade dos dados e a semântica do conjunto são importantes, como coordenadas geográficas, registros estáticos, dados de configuração e retorno de funções.

```python
coordenadas = (10, 20)
print(coordenadas[0])
print(coordenadas[1])
print(len(coordenadas))
# coordenadas[0] = 15  # gera erro, porque tuplas são imutáveis
```

### Propriedades relevantes
* Imutabilidade;
* Ordem preservada;
* Permite elementos repetidos;
* Acesso por índice similar ao de listas;
* Uso apropriado para dados fixos ou sem intenção de alteração.

### Comandos e operações de tuplas
```python
tupla = ("Python", "Java", "C")
print(tupla[0])
print(tupla[1:3])
print(tupla.count("Python"))
print(tupla.index("Java"))
```

## Dicionários
O dicionário é uma estrutura baseada em associação de chave-valor. Diferentemente de listas e tuplas, o dicionário não organiza os elementos por índice, mas por chaves únicas que identificam cada valor. Em Python, a representação é feita com chaves: dicionario = {"chave": "valor"}.

Essa estrutura permite buscas rápidas, atualização de valores e armazenamento semântico de informações. Em termos computacionais, os dicionários apresentam alta eficiência para consultas por chave, sendo muito utilizados em sistemas de cadastro, manipulação de JSON, tabelas de símbolos e processamento de dados em memória.

```python
aluno = {"nome": "Maria", "idade": 20, "curso": "Engenharia"}
print(aluno["nome"])
aluno["idade"] = 21
aluno["periodo"] = 2
print(aluno)
print(aluno.keys())
print(aluno.values())
print(aluno.items())
print("nome" in aluno)
del aluno["periodo"]
print(aluno)
```

### Características principais
* Chave única para cada valor;
* Acesso por chave, não por índice;
* Estrutura mutável;
* Excelente para representar entidades e registros;
* Permite armazenamento de dados heterogêneos.

### Comandos e operações de dicionários
```python
cadastro = {"nome": "Lucas", "idade": 22, "cidade": "Campinas"}
cadastro["curso"] = "ADS"
cadastro["idade"] = 23
print(cadastro.get("nome"))
print(cadastro.get("telefone", "Não informado"))
print(cadastro.pop("cidade"))
print(cadastro)
```

## Conjuntos
Um conjunto é uma coleção de elementos únicos, sem ordem definida e sem índices. Em Python, a sintaxe é apresentada com chaves: nome_conjunto = {valor1, valor2, valor3}. O conjunto é especialmente útil em operações de teoria dos conjuntos, como união, interseção, diferença e verificação de pertinência.

A principal característica do conjunto é a eliminação automática de duplicatas. Além disso, sua estrutura favorece operações lógicas e comparativas entre coleções de valores.

```python
numeros = {1, 2, 2, 3, 3, 4}
print(numeros)
print(2 in numeros)
print(len(numeros))

A = {1, 2, 3}
B = {3, 4, 5}
print(A | B)   # união
print(A & B)   # interseção
print(A - B)   # diferença
print(A ^ B)   # diferença simétrica
```

### Aplicações comuns
* Eliminação de duplicatas;
* Verificação de pertinência;
* Operações de conjunto (união, interseção, diferença);
* Uso em algoritmos de filtragem e deduplicação.

## Comparação entre estruturas de dados
A escolha da estrutura adequada depende do problema e das operações que serão realizadas.

* Lista: mutável, ordenada, indexada, aceita elementos repetidos.
* Tupla: imutável, ordenada, indexada, aceita elementos repetidos.
* Dicionário: mutável, acessado por chave, não usa índice, armazena pares chave-valor.
* Conjunto: mutável, sem ordem, sem índice, armazena elementos únicos.

Essas diferenças são fundamentais para a modelagem de dados e para a construção de algoritmos eficientes.

## Exercício acadêmico
Elabore uma estrutura para representar alunos de uma turma, armazenando nome, idade, curso e média final. Em seguida, percorra os dados e exiba a relação dos alunos em ordem alfabética por nome.

```python
alunos = [
    {"nome": "João", "idade": 19, "curso": "ADS", "media": 8.7},
    {"nome": "Ana", "idade": 20, "curso": "Engenharia", "media": 9.2},
    {"nome": "Pedro", "idade": 21, "curso": "Direito", "media": 7.8}
]

alunos_ordenados = sorted(alunos, key=lambda aluno: aluno["nome"])

for aluno in alunos_ordenados:
    print(f"Nome: {aluno['nome']} | Idade: {aluno['idade']} | Curso: {aluno['curso']} | Média: {aluno['media']}")
```

# Modelagem de Banco de Dados

## Objetivo da aula

Nesta aula, vamos compreender como representar o mundo real em um banco de dados de forma organizada, clara e funcional. A modelagem é a etapa inicial de qualquer projeto de dados e permite definir quais informações serão armazenadas, como elas se relacionam e como serão implementadas depois em um sistema.

## 1. Abstração

A abstração é a capacidade de identificar apenas os elementos relevantes de um problema e ignorar detalhes que não afetam a solução final.

Em banco de dados, isso significa escolher as informações que realmente importam para o sistema e deixar de lado o que não será armazenado ou consultado.

### Exemplo
Em um sistema de biblioteca, não precisamos registrar a cor da cadeira na sala de leitura para controlar empréstimos. O que importa são dados como:
- aluno
- livro
- data do empréstimo
- data de devolução
- status da entrega

### Importância da abstração
- reduz a complexidade do problema
- facilita a análise do sistema
- ajuda a criar um modelo mais correto
- evita dados desnecessários e inconsistentes

## 2. Níveis de abstração de banco de dados

### 2.1 Nível conceitual
Representa o problema do ponto de vista do negócio, sem preocupação com tecnologia. É a visão mais geral do sistema.

Exemplo: "um aluno pode fazer vários empréstimos de livros".

### 2.2 Nível lógico
Organiza as informações em estruturas como tabelas, atributos e relacionamentos.

Exemplo: identificar que a tabela Aluno está relacionada à tabela Empréstimo.

### 2.3 Nível físico
Define como os dados serão armazenados no banco, incluindo regras de armazenamento, tipos de dados, índices e estrutura real do banco.

Exemplo: usar INT, VARCHAR, DATE, FK e PK.

### Fluxo geral
Mundo real -> nível conceitual -> nível lógico -> nível físico

## 3. Modelo Entidade-Relacionamento (MER)

O MER foi proposto por Peter Chen e é amplamente usado para modelar sistemas de informação. Ele representa a realidade por meio de entidades, atributos e relacionamentos.



## 4. Entidades

Uma entidade representa um objeto, conceito ou evento do mundo real que pode ter informações armazenadas.

### Exemplos
- Aluno
- Livro
- Professor
- Cliente
- Pedido
- Produto

### Entidade forte
É aquela que existe independentemente de outra entidade.

Exemplo: Aluno, Livro, Produto.

### Entidade fraca
É aquela que depende da existência de outra entidade para ter sentido.

Exemplo: Dependência financeira de um cliente, ou emissão de uma nota que depende de um pedido.

### Representação
- Retângulo
- Nome da entidade dentro do retângulo

## 5. Atributos

Atributos são as características que descrevem a entidade.

### Exemplos
- Aluno: nome, matrícula, email, telefone
- Livro: código, título, autor, preço
- Pedido: número, data, valor total

### Tipos de atributos

#### Simples
Armazena apenas um valor.

Exemplo: nome do aluno.

#### Composto
É formado por mais de um valor relacionado.

Exemplo: endereço = rua, bairro, cidade, CEP.

#### Derivado ou calculado
É obtido a partir de outros atributos.

Exemplo: idade, calculada a partir da data de nascimento.

#### Multivalorado
Pode guardar mais de um valor do mesmo tipo.

Exemplo: telefones de contato do cliente.

#### Identificador ou determinante
Serve para identificar unicamente uma ocorrência da entidade.

Exemplo: matrícula de aluno, código de produto.

## 6. Relacionamentos

Relacionamento é a ligação entre duas ou mais entidades. Ele representa a associação existente entre os dados.

### Exemplos
- Aluno realiza matrícula em disciplina
- Cliente realiza pedido
- Professor ministra disciplina
- Produto está em categoria

### Representação
- Losango

### Graus de relacionamento
- Unário: relacionamento dentro da mesma entidade
- Binário: relacionamento entre duas entidades
- Ternário: relacionamento entre três entidades
- N-ário: relacionamento entre várias entidades

### Exemplo simples
- Um aluno pode fazer vários pedidos
- Um pedido pertence a um único aluno

Esse é um relacionamento do tipo 1:N

## 7. Cardinalidade do relacionamento

A cardinalidade descreve quantas instâncias de uma entidade podem se relacionar com outra.

### Tipos principais
- 1 para 1 (1:1)
- 1 para muitos (1:N)
- muitos para muitos (N:N)

### Exemplos
- Um país tem uma capital: 1:1
- Um cliente pode fazer vários pedidos: 1:N
- Um aluno pode cursar várias disciplinas e uma disciplina pode ter vários alunos: N:N

## 8. Participação do relacionamento

A participação indica se a entidade precisa obrigatoriamente participar do relacionamento.

### Tipos
- Total: obrigatória
- Parcial: opcional

### Exemplo
- Em um sistema de escola, um aluno pode ser matriculado em disciplina obrigatoriamente? Em geral, a matrícula é obrigatória para que exista um vínculo entre aluno e disciplina.
- Já um professor pode não estar associado a nenhuma disciplina em certo período.

## 9. Diagrama Entidade-Relacionamento (DER)

O Diagrama ER é a representação gráfica do modelo. Ele reúne entidades, atributos e relacionamentos em uma única visão.

Esse modelo serve de base para:
- a análise do sistema
- a criação do modelo lógico
- a implementação em banco de dados
- a comunicação entre analistas e desenvolvedores

## 10. Como interpretar um modelo

Ao analisar um problema, siga esta sequência:

1. identificar as entidades
2. listar os atributos
3. definir os relacionamentos
4. estabelecer a cardinalidade
5. verificar a participação
6. desenhar o DER
7. transformar em modelo lógico
8. gerar as tabelas em SQL



## MER = MODELO ENTIDADE E RELACIONAMENTO
* Modelo criado por Peter Chen em 1976, e teve como base a teoria relacional.
* Segundo Chen, a visão de uma dada realidade, baseia-se no relacionamento entre conceitos desta realidade, os quais retratam os fatos que governam esta mesma realidade, e que cada conceito (entidade ou relacionamento) pode possuir atributos (qualificadores desta realidade).
### Entidade
* É uma abstração de um objeto do mundo real.
* É Qualquer coisa para a qual desejamos guardar informação.
* Conjunto de objetos individuais chamados de instâncias.
* Uma instância deve ter uma identidade distinta de todas as outras.
* Representaca0: um retangulo com o nome dentro

Tipos de atributos: 
- Simples - Este atributo armazena aoenas um valor.

- Derivado/Calculado - ESte atributo depende de outro atributo para existir. - Representacao: traco com elipse com asterisco
- Multivalorado - Este atributo armazena mais de um valor do mesmo tipo.
- Identificador/Determinante - Este atributo identifica unicamente uma ocorrencia no banco de dados - Representacao: traco com linha cruzando e elipse cheia

### Relacionamento
* E a ligacao entre duas ou mais entidades
* Representado por um losangulo

Graus de relacionamentos:
- Unario:
- Binario:
- Ternario:
- N-ario

### CARDINALIDADE DO RELACIONAMENTO

CONCEITO:

* É a propriedade do relacionamento que define exatamente com quantas instâncias de uma entidade pode se relacionar com outras entidades.
* É a quantidade de vezes que uma entidade pode participar da outra entidade.

PODE SER:

CONCEITO:

## Exemplo prático: Sistema de biblioteca

Vamos modelar um sistema simples de biblioteca para entender melhor o uso do MER.

### Entidades
- Aluno
- Livro
- Autor
- Empréstimo

### Atributos
#### Aluno
- matricula
- nome
- telefone
- email

#### Livro
- codigo
- titulo
- ano_publicacao
- genero

#### Autor
- codigo_autor
- nome
- nacionalidade

#### Empréstimo
- codigo_emprestimo
- data_emprestimo
- data_devolucao
- status

### Relacionamentos
- Aluno realiza Empréstimo
- Livro participa de Empréstimo
- Autor escreve Livro

### Cardinalidade
- Um aluno pode realizar muitos empréstimos; cada empréstimo pertence a um aluno.
  - Cardinalidade: 1:N
- Um livro pode aparecer em vários empréstimos; cada empréstimo refere-se a um livro.
  - Cardinalidade: 1:N
- Um autor pode escrever vários livros; um livro tem um ou mais autores.
  - Cardinalidade: N:N

### Interpretação do modelo
- A entidade Aluno é forte, pois pode existir sem depender de outra entidade.
- A entidade Empréstimo depende de Aluno e Livro para existir, então é um exemplo de relacionamento importante entre outras entidades.
- O atributo matricula de Aluno funciona como identificador único.
- O código do livro é o identificador da entidade Livro.

## 12. Exemplo de relacionamento N:N

Um relacionamento muitos-para-muitos ocorre quando várias ocorrências de uma entidade se relacionam com várias ocorrências de outra.

### Exemplo
- Aluno e Disciplina
- Cliente e Produto
- Professor e Disciplina

#### Situação
Um aluno pode se matricular em várias disciplinas, e uma disciplina pode ter vários alunos.

Nesse caso, o relacionamento não pode ser armazenado diretamente em uma única tabela sem criar uma tabela associativa.

### Tabela associativa
A tabela associativa armazena a relação entre as entidades.

Exemplo:

| id_matricula | id_aluno | id_disciplina | data_matricula |
|-------------|----------|---------------|----------------|
| 1           | 10       | 2             | 2026-08-10     |
| 2           | 10       | 5             | 2026-08-12     |

Esse tipo de tabela é necessário para transformar N:N em estruturas que o banco relacional consegue representar.

## 15. Exemplos visuais de relacionamento

### Exemplo 1: relacionamento 1:N

```text
┌──────────────┐        realiza        ┌──────────────┐
│    ALUNO     │ ----------------------> │  EMPRÉSTIMO   │
├──────────────┤                        ├──────────────┤
│ matricula    │                        │ codigo_emp    │
│ nome         │                        │ data_emp      │
│ telefone     │                        │ status        │
└──────────────┘                        └──────────────┘
       ^                                           |
       |                                           |
       | 1                                         N
       |                                           |
       └──────────────────────────────────────────┘
```

Interpretação:
- Um aluno pode realizar vários empréstimos.
- Cada empréstimo pertence a um único aluno.
- Cardinalidade: 1:N

### Exemplo 2: relacionamento N:N

```text
┌──────────────┐       matricula      ┌──────────────┐
│    ALUNO     │ --------------------> │  DISCIPLINA   │
├──────────────┤                       ├──────────────┤
│ id_aluno     │                       │ id_disciplina │
│ nome         │                       │ nome          │
│ email        │                       │ professor     │
└──────────────┘                       └──────────────┘
       ^                                      ^
       |                                      |
       | N                                    N
       |                                      |
       └─────────────── tabela associativa ──┘

             ┌───────────────────────────┐
             │        MATRICULA          │
             ├───────────────────────────┤
             │ id_matricula (PK)         │
             │ id_aluno (FK)             │
             │ id_disciplina (FK)        │
             │ data_matricula            │
             └───────────────────────────┘
```

Interpretação:
- Um aluno pode estar em várias disciplinas.
- Uma disciplina pode ter vários alunos.
- Para representar isso, utiliza-se uma tabela associativa.
- Cardinalidade: N:N

### Exemplo 3: relacionamento 1:1

```text
┌──────────────┐
│   CLIENTE    │
├──────────────┤
│ id_cliente   │
│ nome         │
│ email        │
└──────┬───────┘
       │ 1
       │
       │
       v
┌──────────────┐
│   LOGIN      │
├──────────────┤
│ id_login     │
│ usuario      │
│ senha        │
└──────────────┘
```

Interpretação:
- Um cliente pode ter um único login.
- Um login pertence a um único cliente.
- Cardinalidade: 1:1

### Fluxo visual do MER para SQL

```text
ENTIDADES + ATRIBUTOS + RELACIONAMENTOS
                ↓
          DIAGRAMA ER
                ↓
         MODELO LÓGICO
                ↓
      TABELAS + CHAVES + RELAÇÕES
                ↓
              SQL
```

Esse tipo de representação ajuda a visualizar a transição do mundo real para a estrutura do banco de dados.

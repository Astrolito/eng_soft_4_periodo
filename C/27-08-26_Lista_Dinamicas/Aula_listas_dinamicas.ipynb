# Estruturas Dinamicas
## Listas dinamicas
* Uma forma de alterar o valor que a memoria consegue comportar, atraves de nós
* Alocacao dinamica dos tamanhos dos vetores
* NÓ - ele amarra o ultimo elemento de memoria ao proximo elemento de memoria atraves de ponteiros, 
* Endereço de memoria tem o dado e o endereço do proximo dado (ponteiro,)


````c
#include <stdio.h>
#include <stdlib.h>

typedef struct NO{
	int dado;
	struct NO *prox;
}Elemento;

struct Lista{
	Elemento *inicio;
};

struct Lista* inicializar(){
	struct Lista *lista = (struct Lista*) malloc(sizeof(struct Lista));
	lista->inicio = NULL;
	return lista;
}

Elemento* criarNO(int valor){
	Elemento *novo = (Elemento*) malloc(sizeof(Elemento));
	if(novo == NULL){
		printf("Especo insuficiente!\n");
		system("exit");
	}else{
		novo->dado = valor;
		novo->prox = NULL;
		return novo;	
	}
}

Elemento* buscar(struct Lista *lista, int valorProcurado){
	Elemento *atual = lista->inicio;
	while(atual != NULL){
		if(atual->dado == valorProcurado){
			return atual;
		}
		atual = atual->prox;
	}
	return atual;
}

void inserirFim(struct Lista *lista, Elemento *novo){
	if(lista->inicio == NULL){
		lista->inicio = novo;
	}else{
		Elemento *atual = lista->inicio;
		while(atual->prox != NULL){
			atual = atual->prox;
		}
		atual->prox = novo;
	}
	printf("Inserido com sucesso!\n");
}

void inserirInicio(struct Lista *lista, Elemento *novo){
	novo->prox = lista->inicio;
	lista->inicio = novo;
	printf("Inserido com sucesso!\n");
}

void inserirMeio(struct Lista *lista, Elemento *novo, int valorProcurado){
	Elemento *posicao = buscar(lista, valorProcurado);
	if(posicao != NULL){
		novo->prox = posicao->prox;
		posicao->prox = novo;
		printf("Inserido com sucesso!\n");
	}else{
		inserirFim(lista, novo);
	}
}

void exibirLista(struct Lista *lista){
	if(lista->inicio == NULL){
		printf("Lisa esta vazia!\n");
	}else{
		Elemento *atual = lista->inicio;
		printf("Lista: ");
		do{
			printf("%d, ", atual->dado);
			atual = atual->prox;
		}while(atual != NULL);
		printf("\n");
	}
}

void removerInicio(struct Lista *lista){
	if(lista->inicio == NULL){
		printf("Nao ha elementos a remover!\n");
	}else{
		Elemento *noRemovido = lista->inicio;
		lista->inicio = noRemovido->prox;
		printf("No removido: %d\n", noRemovido->dado);
		free(noRemovido);
	}
}

void removerFim(struct Lista *lista){
	if(lista->inicio == NULL){
		printf("Nao ha elementos a remover!\n");
	}else{
		Elemento *atual = lista->inicio;
		Elemento *antecessor = NULL;
		while(atual->prox != NULL){
			antecessor = atual;
			atual = atual->prox;
		}
		if(antecessor != NULL){
			antecessor->prox = NULL;
		}
		printf("Elemento removido: %d\n", atual->dado);
		free(atual);
	}
}

void removerMeio(struct Lista *lista, int valorProcurado){
	if(lista->inicio == NULL){
		printf("Nao ha elementos a remover!\n");
	}else{
		Elemento *atual = lista->inicio;
		Elemento *antecessor = NULL;
		do{
			if(atual->dado == valorProcurado){
				break;
			}else{
				antecessor = atual;
				atual = atual->prox;
			}
		}while(atual != NULL);
		
		if(atual == NULL){
			printf("Elemetno nao encontrado!\n");
		}else{
			if(antecessor != NULL){
				antecessor->prox = atual->prox;
			}
			printf("Elemento removido: %d\n", atual->dado);
			free(atual);
		}
		
	}
}




int main(){
	
	struct Lista *lista;
	
	lista = inicializar();
	
	inserirFim(lista, criarNO(5));
	inserirFim(lista, criarNO(10));
	inserirFim(lista, criarNO(15));
	inserirFim(lista, criarNO(20));
	inserirFim(lista, criarNO(25));
	inserirFim(lista, criarNO(30));
	inserirFim(lista, criarNO(35));
	inserirFim(lista, criarNO(40));
	inserirFim(lista, criarNO(45));
	
	exibirLista(lista);
	
	removerFim(lista);
	
	exibirLista(lista);
	
	removerInicio(lista);
	
	exibirLista(lista);
	
	removerMeio(lista, 30);
	
	exibirLista(lista);
	
	removerMeio(lista, 30);
	
	exibirLista(lista);
	
	
	
	return 0;
}



````

# Explicacao do codigo: lista dinamica

## 1. Bibliotecas

```c
#include <stdio.h>
#include <stdlib.h>
```

- `stdio.h` serve para usar funcoes como `printf` e entrada/saida.
- `stdlib.h` serve para usar `malloc`, `free` e funcoes de alocacao dinamica.

### Exemplo pratico

Quando o programa quer mostrar uma mensagem e criar um novo no, ele usa essas bibliotecas:

```c
printf("Lista criada com sucesso!\n");
Elemento *novo = (Elemento*) malloc(sizeof(Elemento));
```

---

## 2. Estrutura do no

```c
typedef struct NO{
    int dado;
    struct NO *prox;
} Elemento;
```

Cada no guarda:

- `dado`: o valor do elemento
- `prox`: o endereco do proximo no

A lista funciona porque cada no aponta para o seguinte:

```text
[5] -> [10] -> [15] -> NULL
```

### Exemplo pratico

Se criarmos um no com valor 20, ele seria algo assim:

```text
[ dado: 20 | prox: ... ]
```

O campo `prox` pode apontar para outro no ou para `NULL`, que indica o fim da lista.

---

## 3. Estrutura da lista

```c
struct Lista{
    Elemento *inicio;
};
```

A lista tem apenas um ponteiro chamado `inicio`, que aponta para o primeiro elemento.

### Exemplo pratico

Quando a lista ainda esta vazia:

```text
inicio -> NULL
```

Quando ela ja possui elementos:

```text
inicio -> [5] -> [10] -> [15] -> NULL
```

---

## 4. Inicializacao da lista

```c
struct Lista* inicializar(){
    struct Lista *lista = (struct Lista*) malloc(sizeof(struct Lista));
    lista->inicio = NULL;
    return lista;
}
```

Essa funcao cria a estrutura da lista e define `inicio` como `NULL` para indicar que ela comeca vazia.

### Exemplo pratico

```c
struct Lista *lista;
lista = inicializar();
```

Resultado:

```text
lista->inicio = NULL
```

Isso significa que a lista ainda nao tem nenhum elemento.

---

## 5. Criacao de um no

```c
Elemento* criarNO(int valor){
    Elemento *novo = (Elemento*) malloc(sizeof(Elemento));
    if(novo == NULL){
        printf("Especo insuficiente!\n");
        system("exit");
    }else{
        novo->dado = valor;
        novo->prox = NULL;
        return novo;
    }
}
```

A funcao cria um novo no na memoria, guarda o valor e aponta `prox` para `NULL`.

### Exemplo pratico

```c
Elemento *novo = criarNO(25);
```

Esse no fica assim:

```text
[ dado: 25 | prox: NULL ]
```

Depois ele pode ser encadeado na lista.

---

## 6. Busca por valor

```c
Elemento* buscar(struct Lista *lista, int valorProcurado){
    Elemento *atual = lista->inicio;
    while(atual != NULL){
        if(atual->dado == valorProcurado){
            return atual;
        }
        atual = atual->prox;
    }
    return atual;
}
```

Essa funcao percorre a lista do inicio ao fim comparando cada valor com o valor procurado.

### Exemplo pratico

Lista atual:

```text
5 -> 10 -> 15 -> 20 -> NULL
```

Se executarmos:

```c
Elemento *resultado = buscar(lista, 15);
```

A funcao retorna o endereco do no que tem o valor `15`.

Se executarmos:

```c
Elemento *resultado = buscar(lista, 99);
```

Como `99` nao existe, a funcao retorna `NULL`.

---

## 7. Insercao de elementos

### 7.1 Insercao no fim

```c
void inserirFim(struct Lista *lista, Elemento *novo){
    if(lista->inicio == NULL){
        lista->inicio = novo;
    }else{
        Elemento *atual = lista->inicio;
        while(atual->prox != NULL){
            atual = atual->prox;
        }
        atual->prox = novo;
    }
    printf("Inserido com sucesso!\n");
}
```

A funcao anda pela lista ate encontrar o ultimo no e conecta o novo no ao final.

### Exemplo pratico

Antes:

```text
5 -> 10 -> NULL
```

Execucao:

```c
inserirFim(lista, criarNO(15));
```

Depois:

```text
5 -> 10 -> 15 -> NULL
```

### 7.2 Insercao no inicio

```c
void inserirInicio(struct Lista *lista, Elemento *novo){
    novo->prox = lista->inicio;
    lista->inicio = novo;
    printf("Inserido com sucesso!\n");
}
```

A funcao coloca o novo no na frente da lista, fazendo ele passar a ser o primeiro elemento.

### Exemplo pratico

Antes:

```text
5 -> 10 -> 15 -> NULL
```

Execucao:

```c
inserirInicio(lista, criarNO(3));
```

Depois:

```text
3 -> 5 -> 10 -> 15 -> NULL
```

### 7.3 Insercao no meio

```c
void inserirMeio(struct Lista *lista, Elemento *novo, int valorProcurado){
    Elemento *posicao = buscar(lista, valorProcurado);
    if(posicao != NULL){
        novo->prox = posicao->prox;
        posicao->prox = novo;
        printf("Inserido com sucesso!\n");
    }else{
        inserirFim(lista, novo);
    }
}
```

A funcao procura o valor desejado e insere o novo no imediatamente depois dele.

### Exemplo pratico

Antes:

```text
5 -> 10 -> 15 -> NULL
```

Execucao:

```c
inserirMeio(lista, criarNO(12), 10);
```

Depois:

```text
5 -> 10 -> 12 -> 15 -> NULL
```

## 8. Exibicao da lista

```c
void exibirLista(struct Lista *lista){
    if(lista->inicio == NULL){
        printf("Lisa esta vazia!\n");
    }else{
        Elemento *atual = lista->inicio;
        printf("Lista: ");
        do{
            printf("%d, ", atual->dado);
            atual = atual->prox;
        }while(atual != NULL);
        printf("\n");
    }
}
```

Essa funcao percorre a lista inteira e imprime cada valor na ordem em que aparecem.

### Exemplo pratico

Se a lista for:

```text
5 -> 10 -> 15 -> NULL
```

A chamada:

```c
exibirLista(lista);
```

imprime:

```text
Lista: 5, 10, 15,
```

## 9. Remocao de elementos

### 9.1 Remocao no inicio

```c
void removerInicio(struct Lista *lista){
    if(lista->inicio == NULL){
        printf("Nao ha elementos a remover!\n");
    }else{
        Elemento *noRemovido = lista->inicio;
        lista->inicio = noRemovido->prox;
        printf("No removido: %d\n", noRemovido->dado);
        free(noRemovido);
    }
}
```

Essa funcao retira o primeiro no da lista e ajusta o ponteiro `inicio` para o proximo elemento.

### Exemplo pratico

Antes:

```text
7 -> 9 -> 13 -> NULL
```

Execucao:

```c
removerInicio(lista);
```

Depois:

```text
9 -> 13 -> NULL
```

### 9.2 Remocao no final

```c
void removerFim(struct Lista *lista){
    if(lista->inicio == NULL){
        printf("Nao ha elementos a remover!\n");
    }else{
        Elemento *atual = lista->inicio;
        Elemento *antecessor = NULL;
        while(atual->prox != NULL){
            antecessor = atual;
            atual = atual->prox;
        }
        if(antecessor != NULL){
            antecessor->prox = NULL;
        }
        printf("Elemento removido: %d\n", atual->dado);
        free(atual);
    }
}
```

A funcao percorre a lista ate o ultimo no e libera esse elemento da cadeia.

### Exemplo pratico

Antes:

```text
2 -> 4 -> 6 -> 8 -> NULL
```

Execucao:

```c
removerFim(lista);
```

Depois:

```text
2 -> 4 -> 6 -> NULL
```

### 9.3 Remocao no meio

```c
void removerMeio(struct Lista *lista, int valorProcurado){
    if(lista->inicio == NULL){
        printf("Nao ha elementos a remover!\n");
    }else{
        Elemento *atual = lista->inicio;
        Elemento *antecessor = NULL;
        do{
            if(atual->dado == valorProcurado){
                break;
            }else{
                antecessor = atual;
                atual = atual->prox;
            }
        }while(atual != NULL);

        if(atual == NULL){
            printf("Elemetno nao encontrado!\n");
        }else{
            if(antecessor != NULL){
                antecessor->prox = atual->prox;
            }
            printf("Elemento removido: %d\n", atual->dado);
            free(atual);
        }
    }
}
```

Essa funcao busca um valor e remove o no correspondente, reconectando a lista para que ela continue correta.

### Exemplo pratico

Antes:

```text
10 -> 20 -> 30 -> 40 -> NULL
```

Execucao:

```c
removerMeio(lista, 30);
```

Depois:

```text
10 -> 20 -> 40 -> NULL
```

## 10. Funcao `main`

```c
int main(){
    struct Lista *lista;
    lista = inicializar();

    inserirFim(lista, criarNO(5));
    inserirFim(lista, criarNO(10));
    inserirFim(lista, criarNO(15));
    inserirFim(lista, criarNO(20));
    inserirFim(lista, criarNO(25));
    inserirFim(lista, criarNO(30));
    inserirFim(lista, criarNO(35));
    inserirFim(lista, criarNO(40));
    inserirFim(lista, criarNO(45));

    exibirLista(lista);

    removerFim(lista);
    exibirLista(lista);

    removerInicio(lista);
    exibirLista(lista);

    removerMeio(lista, 30);
    exibirLista(lista);

    removerMeio(lista, 30);
    exibirLista(lista);

    return 0;
}
```

Este programa cria uma lista, insere varios valores, remove elementos do fim, do inicio e do meio, e depois mostra a lista em cada etapa.

### Exemplo pratico de execucao

A lista comeca vazia:

```text
inicio -> NULL
```

Depois de inserir os valores 5, 10, 15, 20, 25, 30, 35, 40 e 45:

```text
5 -> 10 -> 15 -> 20 -> 25 -> 30 -> 35 -> 40 -> 45 -> NULL
```

Quando rodar `removerFim(lista)`, o ultimo elemento (45) e removido:

```text
5 -> 10 -> 15 -> 20 -> 25 -> 30 -> 35 -> 40 -> NULL
```

Quando rodar `removerInicio(lista)`, o primeiro elemento (5) e removido:

```text
10 -> 15 -> 20 -> 25 -> 30 -> 35 -> 40 -> NULL
```

Quando rodar `removerMeio(lista, 30)`, o valor 30 sai da lista:

```text
10 -> 15 -> 20 -> 25 -> 35 -> 40 -> NULL
```

Se chama `removerMeio(lista, 30)` outra vez, ele nao encontra mais o valor e imprime uma mensagem de erro.

## Resumo

| Funcao | O que faz | Exemplo pratico |
|---|---|---|
| `inicializar` | Cria a lista vazia | `inicio -> NULL` |
| `criarNO` | Cria um novo no | `criarNO(25)` |
| `buscar` | Procura um valor | `buscar(lista, 15)` |
| `inserirFim` | coloca no final | `5 -> 10 -> 15 -> NULL` |
| `inserirInicio` | coloca no inicio | `3 -> 5 -> 10 -> 15 -> NULL` |
| `inserirMeio` | coloca depois de um valor | `5 -> 10 -> 12 -> 15 -> NULL` |
| `exibirLista` | mostra todos os valores | `Lista: 5, 10, 15,` |
| `removerInicio` | remove o primeiro | `7 -> 9 -> 13 -> NULL` |
| `removerFim` | remove o ultimo | `2 -> 4 -> 6 -> NULL` |
| `removerMeio` | remove um valor especifico | `10 -> 20 -> 40 -> NULL` |

## Conclusao

A lista dinamica e uma estrutura muito util quando nao sabemos quantos elementos vao entrar na memoria. Em vez de usar tamanho fixo, cada elemento aponta para o proximo, permitindo inserir, buscar, exibir e remover itens de forma flexivel.

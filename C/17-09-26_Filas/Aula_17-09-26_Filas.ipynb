# FILAS

## O que é uma fila?

Uma fila é uma estrutura de dados que segue a regra FIFO:

- FIFO = First In, First Out
- Primeiro que entra, primeiro que sai

Isso significa que o elemento que chega primeiro na fila também é o primeiro a ser atendido ou removido.

Imagine uma fila de banco:

- a primeira pessoa da fila é atendida primeiro;
- depois entra a próxima pessoa;
- e assim por diante.

## Como funciona?

Uma fila tem duas partes importantes:

- frente: o primeiro elemento da fila
- traseira: o último elemento da fila

Quando adicionamos um elemento, ele vai para o final da fila. Quando retiramos, removemos o elemento da frente.

## Representação visual

```text
[10] -> [20] -> [30] -> [40]
  ^                         ^
 frente                    traseira
```

Se removermos o primeiro elemento:

```text
[20] -> [30] -> [40]
  ^                 ^
 frente            traseira
```

## Operações básicas

### 1. Enfileirar
Adicionar um elemento no final da fila.

Exemplo:

```text
Enfileira 10 -> [10]
Enfileira 20 -> [10] -> [20]
Enfileira 30 -> [10] -> [20] -> [30]
```

### 2. Desenfileirar
Remover o elemento da frente da fila.

Exemplo:

```text
Antes: [10] -> [20] -> [30]
Depois: [20] -> [30]
```

### 3. Ver a frente
Consultar qual elemento está na frente da fila sem removê-lo.

```text
Fila: [10] -> [20] -> [30]
Frente: 10
```

## Exemplos práticos

### 1. Fila do banco
Uma pessoa chega primeiro e é atendida primeiro.

```text
Cliente A -> Cliente B -> Cliente C
```

A ordem de atendimento é:

```text
A, B, C
```

### 2. Impressão de documentos
Quando várias pessoas mandam impressões, a impressora atende na ordem de chegada.

```text
Documento 1 -> Documento 2 -> Documento 3
```

### 3. Processos do sistema operacional
O computador organiza processos em fila para serem executados em ordem.

```text
Processo 1 -> Processo 2 -> Processo 3
```

### 4. Atendimento em restaurante ou loja
As pessoas entram na fila e são chamadas na ordem de chegada.

```text
Cliente 1 -> Cliente 2 -> Cliente 3 -> Cliente 4
```

## Vantagens da fila

- organiza a ordem de atendimento;
- segue uma lógica simples e previsível;
- é muito útil em sistemas e algoritmos;
- ajuda a controlar recursos compartilhados.

## Desvantagens

- nem sempre é a melhor estrutura para acessos aleatórios;
- para remover do meio da fila não é eficiente;
- exige cuidado com a manipulação dos elementos.

## Resumo

A fila é uma estrutura de dados baseada na regra FIFO:

- o primeiro a entrar é o primeiro a sair;
- serve para organização de atendimento e processamento;
- é muito usada em programação e em sistemas computacionais.

Em C, normalmente implementamos filas com:

- lista encadeada;
- ou vetor com índices de início e fim.

A lógica principal é sempre a mesma: inserir no final e remover na frente.

```text
Enfileira 5
Enfileira 8
Enfileira 12
Desenfileira
```

Resultado:

```text
Antes: [5] -> [8] -> [12]
Depois: [8] -> [12]
```

A fila é muito importante porque representa bem situações do cotidiano em que a ordem de chegada precisa ser respeitada.

## Implementação em C

```c
#include <stdio.h>
#include <stdlib.h>

struct No {
    int dado;
    struct No *prox;
};

typedef struct No No;

typedef struct {
    No *frente;
    No *tras;
} Fila;

void inicializarFila(Fila *fila) {
    fila->frente = NULL;
    fila->tras = NULL;
}

int filaVazia(Fila *fila) {
    return fila->frente == NULL;
}

void enfileirar(Fila *fila, int valor) {
    No *novo = (No *) malloc(sizeof(No));
    novo->dado = valor;
    novo->prox = NULL;

    if (filaVazia(fila)) {
        fila->frente = novo;
        fila->tras = novo;
    } else {
        fila->tras->prox = novo;
        fila->tras = novo;
    }
}

int desenfileirar(Fila *fila) {
    if (filaVazia(fila)) {
        printf("Fila vazia!\n");
        return -1;
    }

    No *removido = fila->frente;
    int valor = removido->dado;
    fila->frente = removido->prox;

    if (fila->frente == NULL) {
        fila->tras = NULL;
    }

    free(removido);
    return valor;
}

void exibirFila(Fila *fila) {
    No *atual = fila->frente;

    printf("Fila: ");
    while (atual != NULL) {
        printf("%d ", atual->dado);
        atual = atual->prox;
    }
    printf("\n");
}

int main() {
    Fila fila;
    inicializarFila(&fila);

    enfileirar(&fila, 10);
    enfileirar(&fila, 20);
    enfileirar(&fila, 30);

    exibirFila(&fila);

    printf("Saiu: %d\n", desenfileirar(&fila));
    exibirFila(&fila);

    return 0;
}
```

## Como essa implementação funciona?

- `enfileirar` coloca o valor no fim da fila;
- `desenfileirar` retira o valor da frente;
- `filaVazia` verifica se a fila está vazia;
- `exibirFila` percorre todos os elementos e mostra a ordem atual.

## Saída esperada

```text
Fila: 10 20 30
Saiu: 10
Fila: 20 30
```

Essa implementação mostra claramente a regra FIFO, que é a base das filas em estruturas de dados.

## Explicação de cada sessão do código

### 1) Bibliotecas

```c
#include <stdio.h>
#include <stdlib.h>
```

Essas bibliotecas são necessárias porque o programa usa:

- `printf` e `scanf` para entrada e saída;
- `malloc` e `free` para alocação e liberação de memória.

### 2) Estrutura do nó

```c
struct No {
    int dado;
    struct No *prox;
};
```

Cada elemento da fila é representado por um nó. Ele guarda:

- `dado`: o valor armazenado;
- `prox`: ponteiro para o próximo nó da fila.

### 3) Estrutura da fila

```c
typedef struct {
    No *frente;
    No *tras;
} Fila;
```

A fila possui dois ponteiros:

- `frente`: aponta para o primeiro elemento;
- `tras`: aponta para o último elemento.

Esses ponteiros ajudam a inserir no fim e remover no início.

### 4) Inicialização da fila

```c
void inicializarFila(Fila *fila) {
    fila->frente = NULL;
    fila->tras = NULL;
}
```

A fila começa vazia. Por isso, tanto a frente quanto a traseira apontam para `NULL`.

### 5) Verificação de fila vazia

```c
int filaVazia(Fila *fila) {
    return fila->frente == NULL;
}
```

Se a frente for `NULL`, a fila está vazia.

### 6) Enfileirar

```c
void enfileirar(Fila *fila, int valor) {
    No *novo = (No *) malloc(sizeof(No));
    novo->dado = valor;
    novo->prox = NULL;

    if (filaVazia(fila)) {
        fila->frente = novo;
        fila->tras = novo;
    } else {
        fila->tras->prox = novo;
        fila->tras = novo;
    }
}
```

Esta função coloca o novo valor no fim da fila:

- se a fila estiver vazia, o novo nó vira o primeiro e o último;
- se não estiver vazia, o último nó atual aponta para o novo nó e a traseira é atualizada.

### 7) Desenfileirar

```c
int desenfileirar(Fila *fila) {
    if (filaVazia(fila)) {
        printf("Fila vazia!\n");
        return -1;
    }

    No *removido = fila->frente;
    int valor = removido->dado;
    fila->frente = removido->prox;

    if (fila->frente == NULL) {
        fila->tras = NULL;
    }

    free(removido);
    return valor;
}
```

Essa função remove o primeiro elemento da fila, ou seja, o que entrou primeiro.

- guarda o valor do primeiro elemento;
- move a frente para o próximo nó;
- se não houver mais elementos, a traseira também é ajustada;
- libera a memória do nó removido.

### 8) Exibir fila

```c
void exibirFila(Fila *fila) {
    No *atual = fila->frente;

    printf("Fila: ");
    while (atual != NULL) {
        printf("%d ", atual->dado);
        atual = atual->prox;
    }
    printf("\n");
}
```

Aqui o programa percorre todos os elementos da fila a partir da frente até o fim.

### 9) Função `main`

```c
int main() {
    Fila fila;
    inicializarFila(&fila);

    enfileirar(&fila, 10);
    enfileirar(&fila, 20);
    enfileirar(&fila, 30);

    exibirFila(&fila);

    printf("Saiu: %d\n", desenfileirar(&fila));
    exibirFila(&fila);

    return 0;
}
```

Nesta parte:

- cria a fila;
- insere 10, 20 e 30;
- mostra a fila na ordem atual;
- remove o elemento da frente;
- mostra novamente a fila após a remoção.

### 10) Saída esperada

```text
Fila: 10 20 30
Saiu: 10
Fila: 20 30
```

Isso mostra que a fila respeita a regra FIFO: o primeiro que entrou saiu primeiro.

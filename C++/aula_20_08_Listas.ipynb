# Listas: conceito e representacao

Uma lista e uma estrutura de dados linear formada por uma sequencia de elementos. Cada elemento ocupa uma posicao logica, normalmente identificada por um indice, e a ordem dos elementos e preservada.

## O que uma lista permite fazer?

- **Inserir:** adicionar um elemento em uma posicao.
- **Remover:** excluir um elemento e reorganizar a lista, quando necessario.
- **Buscar:** localizar um elemento ou sua posicao.
- **Atualizar:** substituir o valor de um elemento existente.
- **Percorrer:** visitar os elementos em ordem para imprimir ou processar seus valores.
- **Ordenar:** reorganizar os elementos de acordo com algum criterio.

## Representacao fisica e representacao logica

A lista e uma ideia abstrata. Ela pode ser implementada de maneiras diferentes:

- Em um **vetor**, os elementos ficam em posicoes contiguas da memoria e o acesso por indice e direto.
- Em uma **lista encadeada**, cada no guarda um valor e um ponteiro para outro no; os elementos nao precisam ocupar enderecos consecutivos.

Neste notebook, o exemplo pratico usa um vetor de tamanho fixo. Portanto, trata-se de uma lista estatica, e nao de uma lista encadeada.

## Complexidade de tempo

A complexidade de tempo descreve como o custo de uma operacao cresce conforme aumenta a quantidade de elementos, representada por `n`. Ela nao e o numero de posicoes do vetor: um vetor com capacidade `10` tem capacidade fixa, mas uma operacao pode acessar uma unica posicao ou percorrer varias.

Na lista estatica deste exemplo:

| Operacao | Complexidade | Motivo |
|---|---:|---|
| Verificar cheia ou vazia | `O(1)` | Consulta apenas `fim`. |
| Inserir no fim | `O(1)` | Escreve na proxima posicao. |
| Remover no fim | `O(1)` | Apenas decrementa `fim`. |
| Imprimir ou percorrer | `O(n)` | Visita todos os elementos. |
| Inserir no inicio ou no meio | `O(n)` | Pode deslocar varios elementos para a direita. |
| Remover no meio | `O(n)` | Pode deslocar varios elementos para a esquerda. |
| Acesso por indice | `O(1)` | O vetor calcula diretamente o endereco da posicao. |

A notacao `O(n)` representa crescimento linear. Assim, inserir no inicio pode ser barato em uma lista pequena, mas exige mais copias quando a lista fica maior.

# Tipos de lista

A forma de armazenamento influencia a capacidade, o acesso e o custo das operacoes.

## 1. Lista estatica: vetor ou array

- Possui capacidade definida antes ou durante a criacao.
- Armazena elementos em memoria contigua.
- Permite acesso direto por indice em `O(1)`.
- Insercoes e remocoes no inicio ou no meio podem exigir deslocamentos em `O(n)`.
- Quando a capacidade chega ao limite, e preciso criar um vetor maior ou recusar a insercao.

## 2. Lista dinamica ou encadeada

- Cresce e diminui durante a execucao com alocacao dinamica.
- Cada no armazena um valor e um ponteiro para o proximo no.
- Nao permite acesso direto eficiente por indice: para chegar a uma posicao, e necessario percorrer os nos.
- Inserir ou remover no inicio e eficiente, desde que o ponteiro do inicio seja conhecido.

## 3. Lista duplamente encadeada

Cada no possui ponteiros para o no anterior e para o proximo. Isso facilita a navegacao nos dois sentidos e a remocao de um no conhecido, mas exige mais memoria e mais atualizacoes de ponteiros.

## 4. Lista circular

O ultimo no aponta novamente para o primeiro. Nao existe um fim representado por `NULL`, por isso o percurso precisa usar uma condicao de parada adequada. Essa estrutura e comum em escalonamento, buffers e tarefas repetitivas; ela nao e infinita, apenas permite retornar ao inicio.

# Operacoes fundamentais

As operacoes abaixo formam a interface mais comum de uma lista. O custo depende da representacao escolhida e da posicao afetada.

- **Insercao:** adiciona um valor. Em um vetor, pode ser necessario deslocar elementos para abrir espaco.
- **Remocao:** exclui um valor. Em um vetor, os elementos posteriores podem ser deslocados para preencher o espaco.
- **Busca:** procura um valor ou uma posicao. Sem uma estrutura auxiliar, a busca por valor costuma ser `O(n)`.
- **Atualizacao:** altera o valor de uma posicao existente; em um vetor, o acesso por indice e `O(1)`.
- **Percurso:** visita os elementos em ordem, geralmente em `O(n)`.
- **Ordenacao:** reorganiza os elementos segundo um criterio; o custo depende do algoritmo utilizado.

Neste exemplo, `inicio` marca a primeira posicao logica e `fim` marca a ultima. A lista esta vazia quando `fim == -1` e possui `fim + 1` elementos.

# Aplicacoes de listas

Listas aparecem como base para estruturas e algoritmos mais especializados:

- **Pilhas:** seguem o principio LIFO, em que o ultimo elemento inserido e o primeiro a sair.
- **Filas:** seguem o principio FIFO, em que o primeiro elemento inserido e o primeiro a sair.
- **Grafos:** listas de adjacencia representam os vizinhos de cada vertice.
- **Arvores:** nos podem manter listas de filhos.
- **Buffers:** armazenam temporariamente dados produzidos e consumidos em ritmos diferentes.
- **Gerenciamento de tarefas:** uma lista pode manter tarefas pendentes, prontas ou concluidas.

A escolha entre vetor e encadeamento depende principalmente da necessidade de acesso por indice, do numero de insercoes e remocoes e da previsibilidade da capacidade.

# Exemplo pratico: lista estatica em C

O programa a seguir usa um vetor para representar uma lista de inteiros com capacidade maxima de cinco elementos. A variavel `fim` controla quantas posicoes estao ocupadas:

```text
capacidade = 5
indices validos = 0, 1, 2, 3, 4
quantidade de elementos = fim + 1
```

A lista nao apaga fisicamente um valor quando ele e removido. Ela apenas atualiza `fim`, ou desloca os valores para manter a sequencia logica. Por isso, somente as posicoes entre `inicio` e `fim` devem ser consideradas validas.

O codigo tambem demonstra as operacoes de inserir no fim, no inicio e em uma posicao intermediaria, alem de remover no fim e em uma posicao informada pelo usuario.

````c
#include <stdio.h>
#include <stdbool.h>
#define tamanho 5

struct Lista{
    int dados[tamanho];
    int inicio, fim;
};

struct Lista lista;

void inicializacao(){
    lista.inicio = 0;
    lista.fim = -1;
}

bool listaCheia(){
    return (lista.fim == tamanho - 1);
}

bool listaVazia(){
    return (lista.fim == -1);
}

void inserirFim(int valor){
    if(listaCheia()){
        printf("Lista cheia!\n");
    } else {
        lista.fim++;
        lista.dados[lista.fim] = valor;
    }
}

void inserirMeio(int valor, int posicao){
    if(listaCheia()){
        printf("Lista cheia!\n");
    } else {
        if(lista.fim >= posicao){
            int auxPosicao = lista.fim;
            do{
                lista.dados[auxPosicao + 1] = lista.dados[auxPosicao];
            }while(--auxPosicao >= posicao); //decremento antres de verificar a condição quando --auxPosicao > posicao se auxPosicao-- for igual a posicao, o loop termina e não é feito o deslocamento do elemento da posição posicao
            lista.dados[posicao] = valor;
            lista.fim++;
            printf("Valor %d inserido na posição %d da lista.\n", valor, posicao);
        }else{
            inserirFim(valor);
        }
    }   
}

void imprimirLista(){
    if(listaVazia()){
        printf("Lista vazia!\n");
    } else {
        int auxPosicao = lista.inicio;
        printf("Lista: ");
        do{
            printf("%d ", lista.dados[auxPosicao]);
        }while(++auxPosicao <= lista.fim);
    }
}

void inserirInicio(int valor){
    if(listaCheia()){
        printf("Lista cheia!\n");
    } else {
        int auxPosicao = lista.fim;
        while(auxPosicao >= 0){
            lista.dados[auxPosicao + 1] = lista.dados[auxPosicao];
            auxPosicao--;
        }
        lista.dados[0] = valor;
        lista.fim++;
        printf("Valor %d inserido na posição 0 da lista.\n", valor);
    }
}


void removerFim(){
    if(listaVazia()){
        printf("Lista vazia!\n");
    } else {
        lista.fim--;
    }
}

void removerMeio(){
    if(listaVazia()){
        printf("Lista vazia!\n");
    } else {
        int posicao;
        printf("Digite a posição do elemento a ser removido: ");
        scanf("%d", &posicao);
        if(posicao < 0 || posicao > lista.fim){
            printf("Posição inválida!\n");
        } else {
            int auxPosicao = posicao;
            while(auxPosicao < lista.fim){
                lista.dados[auxPosicao] = lista.dados[auxPosicao + 1];
                auxPosicao++;
            }
            lista.fim--;
            printf("Elemento da posição %d removido da lista.\n", posicao);
        }
    }
}
int main(){
    inicializacao();
    imprimirLista();
    printf("\n");
    inserirFim(1);
    imprimirLista();
    printf("\n");
    inserirFim(2);
    inserirFim(3);
    imprimirLista();
    printf("\n");
    inserirMeio(10, 1);
    imprimirLista();
    printf("\n");
    inserirInicio(0);
    imprimirLista();
    printf("\n");
    removerFim();
    imprimirLista();
    printf("\n");
    removerMeio();
    imprimirLista();
    printf("\n");
    return 0;
}
   
````

# Explicacao individual do codigo: lista estatica

Este programa implementa uma lista estatica de inteiros usando um vetor. A capacidade e cinco elementos e nao aumenta durante a execucao. A lista e controlada por dois indices: `inicio`, que marca a primeira posicao usada, e `fim`, que marca a ultima posicao ocupada.

## 1. Bibliotecas e capacidade

```c
#include <stdio.h>
#include <stdbool.h>
#define tamanho 5
```

- `stdio.h` disponibiliza `printf` e `scanf`.
- `stdbool.h` disponibiliza o tipo `bool` e os valores `true` e `false`.
- `tamanho` e uma constante de pre-processamento usada para definir a capacidade do vetor.

As posicoes validas do vetor sao `0` a `4`.

## 2. Estrutura e estado da lista

```c
struct Lista{
    int dados[tamanho];
    int inicio, fim;
};

struct Lista lista;
```

`dados` guarda os inteiros. `inicio` e `fim` representam o intervalo logico ocupado. A variavel global `lista` e compartilhada por todas as funcoes, o que simplifica a aula, embora em programas maiores seja comum passar a lista como parametro.

Para uma lista com tres elementos, o estado pode ser:

```text
posicao:  0    1    2    3    4
 dados:  [10] [20] [30] [  ] [  ]
inicio = 0, fim = 2
```

## 3. Inicializacao

```c
void inicializacao(){
    lista.inicio = 0;
    lista.fim = -1;
}
```

`fim = -1` e o marcador de lista vazia: como nao ha indice valido menor que zero, nenhuma posicao pertence a lista. O vetor nao precisa ser zerado, porque valores fora do intervalo logico nao sao lidos.

## 4. Testes de estado

```c
bool listaCheia(){
    return (lista.fim == tamanho - 1);
}

bool listaVazia(){
    return (lista.fim == -1);
}
```

Com capacidade cinco, a lista esta cheia quando `fim == 4`. Ela esta vazia quando `fim == -1`. Essas verificacoes protegem as operacoes contra acesso fora do vetor e contra remocoes sem elementos.

## 5. Insercao no fim

```c
void inserirFim(int valor){
    if(listaCheia()){
        printf("Lista cheia!\\n");
    } else {
        lista.fim++;
        lista.dados[lista.fim] = valor;
    }
}
```

A funcao primeiro verifica a capacidade. Depois incrementa `fim` para a proxima posicao e grava `valor`. Como nenhum elemento precisa ser deslocado, a operacao custa `O(1)`.

## 6. Insercao em uma posicao

```c
void inserirMeio(int valor, int posicao){
    if(listaCheia()){
        printf("Lista cheia!\\n");
    } else {
        if(lista.fim >= posicao){
            int auxPosicao = lista.fim;
            do{
                lista.dados[auxPosicao + 1] = lista.dados[auxPosicao];
            }while(--auxPosicao >= posicao);
            lista.dados[posicao] = valor;
            lista.fim++;
        }else{
            inserirFim(valor);
        }
    }
}
```

Quando `posicao` esta dentro da parte ocupada, os elementos sao copiados da direita para a esquerda do vetor. Essa ordem e importante: copiar no sentido contrario poderia sobrescrever valores ainda nao movidos. Depois, o novo valor ocupa a posicao aberta e `fim` aumenta.

Se `posicao` estiver depois do ultimo elemento, o codigo direciona a insercao para `inserirFim`. O custo e `O(n)` no pior caso. Para uso geral, ainda seria recomendavel validar `posicao >= 0` antes de acessar o vetor.

## 7. Insercao no inicio

```c
void inserirInicio(int valor){
    if(listaCheia()){
        printf("Lista cheia!\\n");
    } else {
        int auxPosicao = lista.fim;
        while(auxPosicao >= 0){
            lista.dados[auxPosicao + 1] = lista.dados[auxPosicao];
            auxPosicao--;
        }
        lista.dados[0] = valor;
        lista.fim++;
    }
}
```

A funcao move todos os elementos uma casa para a direita, iniciando pelo ultimo, e grava o novo valor em `dados[0]`. Inserir no inicio pode mover `n` elementos, portanto custa `O(n)`.

## 8. Percurso e impressao

```c
void imprimirLista(){
    if(listaVazia()){
        printf("Lista vazia!\\n");
    } else {
        int auxPosicao = lista.inicio;
        printf("Lista: ");
        do{
            printf("%d ", lista.dados[auxPosicao]);
        }while(++auxPosicao <= lista.fim);
    }
}
```

A funcao percorre somente o intervalo logico de `inicio` ate `fim`. O `do...while` garante uma impressao quando existe pelo menos um elemento. O percurso custa `O(n)`.

## 9. Remocao no fim

```c
void removerFim(){
    if(listaVazia()){
        printf("Lista vazia!\\n");
    } else {
        lista.fim--;
    }
}
```

Remover o ultimo elemento exige apenas decrementar `fim`, sem deslocamentos. O valor antigo pode continuar no vetor, mas deixou de fazer parte da lista. A operacao custa `O(1)`.

## 10. Remocao em uma posicao

```c
void removerMeio(){
    if(listaVazia()){
        printf("Lista vazia!\\n");
    } else {
        int posicao;
        printf("Digite a posição do elemento a ser removido: ");
        scanf("%d", &posicao);
        if(posicao < 0 || posicao > lista.fim){
            printf("Posição inválida!\\n");
        } else {
            int auxPosicao = posicao;
            while(auxPosicao < lista.fim){
                lista.dados[auxPosicao] = lista.dados[auxPosicao + 1];
                auxPosicao++;
            }
            lista.fim--;
        }
    }
}
```

A posicao precisa estar entre `0` e `fim`. Depois da validacao, os elementos posteriores sao deslocados uma casa para a esquerda, preenchendo o espaco removido. Por isso, o pior caso custa `O(n)`.

O exemplo pressupoe que o usuario digita um inteiro valido. Em uma versao mais robusta, o retorno de `scanf` tambem deve ser verificado para tratar entradas que nao sejam numericas.

## 11. Fluxo da funcao `main`

```c
int main(){
    inicializacao();
    imprimirLista();
    inserirFim(1);
    imprimirLista();
    inserirFim(2);
    inserirFim(3);
    imprimirLista();
    inserirMeio(10, 1);
    imprimirLista();
    inserirInicio(0);
    imprimirLista();
    removerFim();
    imprimirLista();
    removerMeio();
    imprimirLista();
    return 0;
}
```

A sequencia dos estados logicos e:

1. Lista vazia: `[]`.
2. Insercao de `1`: `[1]`.
3. Insercoes de `2` e `3`: `[1, 2, 3]`.
4. Insercao de `10` na posicao `1`: `[1, 10, 2, 3]`.
5. Insercao de `0` no inicio: `[0, 1, 10, 2, 3]`.
6. Remocao no fim: `[0, 1, 10, 2]`.
7. Remocao na posicao escolhida pelo usuario.

As quebras de linha usadas entre as chamadas tornam a saida mais legivel. O programa termina com `return 0`, indicando execucao normal.

## 12. Invariantes importantes

Durante a execucao, o programa deve manter estas regras:

- `inicio` permanece em `0` neste exemplo.
- `-1 <= fim < tamanho`.
- As posicoes de `0` ate `fim` contem os elementos validos.
- A quantidade de elementos e `0` quando `fim == -1`; caso contrario, e `fim + 1`.
- Nunca se deve acessar `dados` com indice menor que `0` ou maior que `tamanho - 1`.

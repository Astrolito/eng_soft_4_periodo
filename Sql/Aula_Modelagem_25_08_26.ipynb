## ABSTRACAO

O conceito de abstração permite ao analista separar da realidade em estudo, as partes que são realmente relevantes para o desenvolvimento do sistema de informações e excluir da modelagem todos os aspectos que não exercem influência sobre o ambiente a ser modelado.
* Exemplo: vamos fazer abstração dos erros e focar nas soluções.

O conceito de abstração consiste em esconder os detalhes de algo, no caso, os detalhes desnecessários.

## MER = MODELO ENTIDADE E RELACIONAMENTO
* Modelo criado por Peter Chen em 1976, e teve como base a teoria relacional.
* Segundo Chen, a visão de uma dada realidade, baseia-se no relacionamento entre conceitos desta realidade, os quais retratam os fatos que governam esta mesma realidade, e que cada conceito (entidade ou relacionamento) pode possuir atributos (qualificadores desta realidade).
### Entidade
* É uma abstração de um objeto do mundo real.
* É Qualquer coisa para a qual desejamos guardar informação.
* Conjunto de objetos individuais chamados de instâncias.
* Uma instância deve ter uma identidade distinta de todas as outras.
* Representaca0: um retangulo com o nome dentro
* entidade forte: entidade que nao precisa de outra entidade para existir
* entidade fraca: so existe uando a forte existir, representado com um retangulo dentro de outro retangulo com nome no meio
### Atributos
* Sao as caracteristicas das entidades, server para qualificar a entidade

Tipos de atributos: 
- Simples - Este atributo armazena aoenas um valor.
- Composto - Este atributo armazena mais de um valor, de tipos diferentes.
- Derivado/Calculado - ESte atributo depende de outro atributo para existir. - Representacao: traco com elipse com asterisco
- Multivalorado - Este atributo armazena mais de um valor do mesmo tipo.
- Identificador/Determinante - Este atributo identifica unicamente uma ocorrencia no banco de dados - Representacao: traco com linha cruzando e elipse cheia

### Relacionamento
* E a ligacao entre duas ou mais entidades
* Representado por um losangulo

Graus de relacionamentos:
- Unario:
- Binario:
- Ternario:
- N-ario

### CARDINALIDADE DO RELACIONAMENTO

CONCEITO:

* É a propriedade do relacionamento que define exatamente com quantas instâncias de uma entidade pode se relacionar com outras entidades.
* É a quantidade de vezes que uma entidade pode participar da outra entidade.

PODE SER:
- um para um (1 x 1);
- um para muitos (1 x N);
- muitos para muitos (N X M):

### PARTICIPAÇÃO DO RELACIONAMENTO

CONCEITO:

* É a propriedade do relacionamento que define a obrigatoriedade ou não de uma entidade participar (se relacionar) de um relacionamento.

PODE SER:

* total (obrigatória) 
* parcial (opcional)

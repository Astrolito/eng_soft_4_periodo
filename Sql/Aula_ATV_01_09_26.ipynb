# Exercício 2 - Modelo de Dados: Matrícula de Alunos em Disciplinas

## Problema
Um aluno pode se matricular em várias disciplinas. Cada disciplina pode ter vários alunos matriculados.

**Identifique as entidades, atributos e o relacionamento com cardinalidades.**

## 1. Identificação das Entidades

### **Entidade 1: ALUNO**
- Representa um estudante que pode se matricular em várias disciplinas

### **Entidade 2: DISCIPLINA**
- Representa uma disciplina que pode ter vários alunos matriculados

## 2. Atributos das Entidades

### **ALUNO**
| Atributo | Tipo | Descrição |
|----------|------|-----------|
| `id_aluno` | INT (PK) | Identificador único do aluno |
| `nome` | VARCHAR(100) | Nome completo do aluno |
| `email` | VARCHAR(100) | Email do aluno |
| `matricula` | VARCHAR(20) (UK) | Número de matrícula único |
| `cpf` | VARCHAR(11) | CPF do aluno |
| `data_nascimento` | DATE | Data de nascimento |
| `telefone` | VARCHAR(15) | Telefone para contato |

### **DISCIPLINA**
| Atributo | Tipo | Descrição |
|----------|------|-----------|
| `id_disciplina` | INT (PK) | Identificador único da disciplina |
| `nome` | VARCHAR(100) | Nome da disciplina |
| `codigo` | VARCHAR(10) (UK) | Código da disciplina (ex: ALC001) |
| `professor` | VARCHAR(100) | Nome do professor responsável |
| `carga_horaria` | INT | Carga horária em horas |
| `periodo` | VARCHAR(20) | Período (1º, 2º semestre) |
| `vagas` | INT | Quantidade de vagas disponíveis |

## 3. Relacionamento e Cardinalidades

### **Tipo de Relacionamento: N:N (Muitos-para-Muitos)**

```
ALUNO (1) ─────────────── (N) DISCIPLINA
    │
    └─── Um aluno pode se matricular em VÁRIAS disciplinas

DISCIPLINA (1) ─────────────── (N) ALUNO
    │
    └─── Uma disciplina pode ter VÁRIOS alunos matriculados
```

### **Cardinalidades:**
- **Aluno → Disciplina:** 1:N (Um aluno para Várias disciplinas)
- **Disciplina → Aluno:** 1:N (Uma disciplina para Vários alunos)
- **Relacionamento Geral:** N:N (Muitos-para-Muitos)

### **Tabela de Junção (Associativa): MATRICULA**

Para implementar um relacionamento N:N, criamos uma tabela intermediária:

| Atributo | Tipo | Descrição |
|----------|------|-----------|
| `id_matricula` | INT (PK) | Identificador único da matrícula |
| `id_aluno` | INT (FK) | Referência ao aluno |
| `id_disciplina` | INT (FK) | Referência à disciplina |
| `data_matricula` | TIMESTAMP | Data em que se matriculou |
| `nota_final` | DECIMAL(5,2) | Nota final do aluno |
| `status` | VARCHAR(20) | Status (Ativo, Concluído, Cancelado) |
| `frequencia` | DECIMAL(5,2) | Percentual de frequência (%)

## 5. Código SQL - Criação das Tabelas

````python
CREATE TABLE aluno (
    id_aluno INT PRIMARY KEY AUTO_INCREMENT,
    nome VARCHAR(100) NOT NULL,
    email VARCHAR(100),
    matricula VARCHAR(20) NOT NULL UNIQUE,
    cpf VARCHAR(11) UNIQUE,
    data_nascimento DATE,
    telefone VARCHAR(15),
    data_inscricao TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);

-- Tabela DISCIPLINA
CREATE TABLE disciplina (
    id_disciplina INT PRIMARY KEY AUTO_INCREMENT,
    nome VARCHAR(100) NOT NULL,
    codigo VARCHAR(10) NOT NULL UNIQUE,
    professor VARCHAR(100),
    carga_horaria INT,
    periodo VARCHAR(20),
    vagas INT
);

-- Tabela de Junção MATRICULA
CREATE TABLE matricula (
    id_matricula INT PRIMARY KEY AUTO_INCREMENT,
    id_aluno INT NOT NULL,
    id_disciplina INT NOT NULL,
    data_matricula TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    nota_final DECIMAL(5,2),
    status VARCHAR(20) DEFAULT 'Ativo',
    frequencia DECIMAL(5,2),
    FOREIGN KEY (id_aluno) REFERENCES aluno(id_aluno),
    FOREIGN KEY (id_disciplina) REFERENCES disciplina(id_disciplina),
    UNIQUE KEY uk_matricula (id_aluno, id_disciplina)
);
````

### **Análise Detalhada - Seção 5: Criação das Tabelas**

#### **Tabela ALUNO**
```sql
CREATE TABLE aluno (
    id_aluno INT PRIMARY KEY AUTO_INCREMENT,
    nome VARCHAR(100) NOT NULL,
    email VARCHAR(100),
    matricula VARCHAR(20) NOT NULL UNIQUE,
    cpf VARCHAR(11) UNIQUE,
    data_nascimento DATE,
    telefone VARCHAR(15),
    data_inscricao TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);
```

**Explicação de cada campo:**

| Campo | Explicação |
|-------|-----------|
| `id_aluno INT PRIMARY KEY AUTO_INCREMENT` | Chave primária única que se incrementa automaticamente. Garante que cada aluno tenha um identificador único |
| `nome VARCHAR(100) NOT NULL` | Campo obrigatório. Armazena o nome completo do aluno com até 100 caracteres |
| `email VARCHAR(100)` | Email opcional (pode ser NULL). Sem constraint UNIQUE pois pode haver alunos sem email |
| `matricula VARCHAR(20) NOT NULL UNIQUE` | Número de matrícula único e obrigatório. Identifica cada aluno na instituição |
| `cpf VARCHAR(11) UNIQUE` | CPF único (opcional). Não pode haver duplicação de CPF no banco |
| `data_nascimento DATE` | Data no formato YYYY-MM-DD. Usada para calcular idade ou gerar relatórios |
| `telefone VARCHAR(15)` | Número de telefone com até 15 caracteres (para incluir códigos de país) |
| `data_inscricao TIMESTAMP` | Registra automaticamente quando o aluno foi cadastrado no sistema |

---

#### **Tabela DISCIPLINA**
```sql
CREATE TABLE disciplina (
    id_disciplina INT PRIMARY KEY AUTO_INCREMENT,
    nome VARCHAR(100) NOT NULL,
    codigo VARCHAR(10) NOT NULL UNIQUE,
    professor VARCHAR(100),
    carga_horaria INT,
    periodo VARCHAR(20),
    vagas INT
);
```

**Explicação de cada campo:**

| Campo | Explicação |
|-------|-----------|
| `id_disciplina INT PRIMARY KEY AUTO_INCREMENT` | Identificador único que incrementa automaticamente |
| `nome VARCHAR(100) NOT NULL` | Nome da disciplina (obrigatório). Ex: "Algoritmos e Programação" |
| `codigo VARCHAR(10) NOT NULL UNIQUE` | Código único para identificação. Ex: "ALC001" |
| `professor VARCHAR(100)` | Nome do professor responsável. Pode ser NULL se não atribuído |
| `carga_horaria INT` | Horas de aula. Ex: 60 horas |
| `periodo VARCHAR(20)` | Semestre em que é oferecida. Ex: "1º semestre" |
| `vagas INT` | Quantidade total de vagas disponíveis na disciplina |

---

#### **Tabela MATRICULA (Associativa)**
```sql
CREATE TABLE matricula (
    id_matricula INT PRIMARY KEY AUTO_INCREMENT,
    id_aluno INT NOT NULL,
    id_disciplina INT NOT NULL,
    data_matricula TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    nota_final DECIMAL(5,2),
    status VARCHAR(20) DEFAULT 'Ativo',
    frequencia DECIMAL(5,2),
    FOREIGN KEY (id_aluno) REFERENCES aluno(id_aluno),
    FOREIGN KEY (id_disciplina) REFERENCES disciplina(id_disciplina),
    UNIQUE KEY uk_matricula (id_aluno, id_disciplina)
);
```

**Explicação de cada campo:**

| Campo | Explicação |
|-------|-----------|
| `id_matricula INT PRIMARY KEY AUTO_INCREMENT` | Identificador único para cada registro de matrícula |
| `id_aluno INT NOT NULL` | Referência (FK) à tabela ALUNO. Obrigatório |
| `id_disciplina INT NOT NULL` | Referência (FK) à tabela DISCIPLINA. Obrigatório |
| `data_matricula TIMESTAMP` | Data/hora que o aluno se matriculou (preenchida automaticamente) |
| `nota_final DECIMAL(5,2)` | Nota final (0.00 até 999.99). Ex: 8.50 |
| `status VARCHAR(20)` | Situação: 'Ativo', 'Concluído', 'Cancelado'. Padrão = 'Ativo' |
| `frequencia DECIMAL(5,2)` | Percentual de frequência (0 até 100.00) |
| `FOREIGN KEY...` | Garante integridade referencial. Aluno e Disciplina devem existir |
| `UNIQUE KEY uk_matricula` | Evita duplicação: um aluno não pode se matricular 2x na mesma disciplina |

**Por que é uma tabela associativa?**
- Resolve o relacionamento N:N entre ALUNO e DISCIPLINA
- Permite armazenar dados adicionais da matrícula (nota, frequência, status)
- Facilita consultas complexas com JOINs

## 7. Exemplos de Consultas SQL

````python
-- CONSULTA 1: Listar todas as disciplinas de um aluno (ex: João Silva - id 1)
SELECT 
    a.nome AS aluno,
    d.codigo,
    d.nome AS disciplina,
    d.professor,
    m.status
FROM disciplina d
INNER JOIN matricula m ON d.id_disciplina = m.id_disciplina
INNER JOIN aluno a ON m.id_aluno = a.id_aluno
WHERE a.id_aluno = 1
ORDER BY d.nome;

-- CONSULTA 2: Listar todos os alunos de uma disciplina (ex: Algoritmos - id 1)
SELECT 
    a.matricula,
    a.nome AS aluno,
    a.email,
    m.data_matricula,
    m.status
FROM aluno a
INNER JOIN matricula m ON a.id_aluno = m.id_aluno
INNER JOIN disciplina d ON m.id_disciplina = d.id_disciplina
WHERE d.id_disciplina = 1
ORDER BY a.nome;

-- CONSULTA 3: Contar quantas disciplinas cada aluno está matriculado
SELECT 
    a.nome,
    COUNT(m.id_disciplina) AS total_disciplinas
FROM aluno a
LEFT JOIN matricula m ON a.id_aluno = m.id_aluno
GROUP BY a.id_aluno, a.nome
ORDER BY total_disciplinas DESC;

-- CONSULTA 4: Contar quantos alunos em cada disciplina
SELECT 
    d.nome AS disciplina,
    d.codigo,
    COUNT(m.id_aluno) AS total_alunos,
    d.vagas,
    (d.vagas - COUNT(m.id_aluno)) AS vagas_disponiveis
FROM disciplina d
LEFT JOIN matricula m ON d.id_disciplina = m.id_disciplina
GROUP BY d.id_disciplina, d.nome, d.codigo, d.vagas
ORDER BY d.nome;

-- CONSULTA 5: Verificar se um aluno está matriculado em uma disciplina específica
SELECT 
    a.nome AS aluno,
    d.nome AS disciplina,
    m.data_matricula,
    m.status
FROM matricula m
INNER JOIN aluno a ON m.id_aluno = a.id_aluno
INNER JOIN disciplina d ON m.id_disciplina = d.id_disciplina
WHERE a.id_aluno = 2 AND d.id_disciplina = 2;
````

```sql
SELECT * FROM aluno a
LEFT JOIN matricula m ON a.id_aluno = m.id_aluno;
-- Resultado: TODOS os alunos (com ou sem matrículas)

SELECT * FROM aluno a

SELECT * FROM aluno a
```

### **Exemplo: Consulta COMPLETA**

```sql
SELECT 
    a.nome,
    COUNT(m.id_disciplina) AS total_disciplinas,
    AVG(m.nota_final) AS media_notas,
    MIN(m.data_matricula) AS primeira_matricula
FROM aluno a
LEFT JOIN matricula m ON a.id_aluno = m.id_aluno
WHERE m.status = 'Ativo'  -- Filtra ANTES de agrupar
GROUP BY a.id_aluno, a.nome  -- Agrupa por aluno
HAVING COUNT(m.id_disciplina) > 0  -- Filtra APÓS agrupar
ORDER BY media_notas DESC  -- Ordena por nota (maior primeiro)
LIMIT 10;  -- Apenas top 10
```

**Passo a passo:**
1. FROM: Pega tabela ALUNO
2. LEFT JOIN: Adiciona matrículas
3. WHERE: Filtra matrículas ativas
4. GROUP BY: Agrupa por aluno
5. HAVING: Filtra grupos (com >0 disciplinas)
6. SELECT: Seleciona e calcula
7. ORDER BY: Ordena resultado
8. LIMIT: Pega apenas 10

### **Cardinalidades:**

- **Projeto → Funcionário:** 1:N (Um projeto para Vários funcionários)
- **Relacionamento Geral:** N:N (Muitos-para-Muitos)

## 5. Código SQL - Criação das Tabelas

````python
-- Tabela FUNCIONÁRIO
CREATE TABLE funcionario (
    id_funcionario INT PRIMARY KEY AUTO_INCREMENT,
    nome VARCHAR(100) NOT NULL,
    email VARCHAR(100),
    cpf VARCHAR(11) NOT NULL UNIQUE,
    departamento VARCHAR(50),
    cargo VARCHAR(50),
    salario DECIMAL(10,2),
    data_admissao DATE,
    status VARCHAR(20) DEFAULT 'Ativo',
    data_cadastro TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);

-- Tabela PROJETO
CREATE TABLE projeto (
    id_projeto INT PRIMARY KEY AUTO_INCREMENT,
    nome VARCHAR(100) NOT NULL,
    descricao TEXT,
    codigo VARCHAR(20) NOT NULL UNIQUE,
    gerente_id INT,
    data_inicio DATE,
    data_fim DATE,
    orcamento DECIMAL(12,2),
    status VARCHAR(20) DEFAULT 'Planejamento',
    data_criacao TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (gerente_id) REFERENCES funcionario(id_funcionario)
);

-- Tabela de Junção ALOCACAO
CREATE TABLE alocacao (
    id_alocacao INT PRIMARY KEY AUTO_INCREMENT,
    id_funcionario INT NOT NULL,
    id_projeto INT NOT NULL,
    data_inicio DATE NOT NULL,
    data_fim DATE,
    funcao_projeto VARCHAR(50),
    horas_alocadas_semana INT,
    status VARCHAR(20) DEFAULT 'Alocado',
    salario_projeto DECIMAL(10,2),
    data_alocacao TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (id_funcionario) REFERENCES funcionario(id_funcionario),
    FOREIGN KEY (id_projeto) REFERENCES projeto(id_projeto),
    UNIQUE KEY uk_alocacao (id_funcionario, id_projeto, data_inicio)
);
````

### **Análise Detalhada - Criação das Tabelas**

#### **Tabela FUNCIONÁRIO**

**Propósito:** Armazenar informações de todos os funcionários da empresa

**Campos críticos:**
- `id_funcionario`: Identificador único (chave primária)
- `cpf`: Único e obrigatório (rastreabilidade legal)
- `email`: Identificação corporativa
- `status`: Controla se funcionário está ativo (importante para folha de pagamento)

**Relacionamentos:**
- É referenciado pela tabela PROJETO (gerente_id)
- É referenciado pela tabela ALOCACAO (funcionários em projetos)

---

#### **Tabela PROJETO**

**Propósito:** Armazenar informações dos projetos da empresa

**Campos críticos:**
- `id_projeto`: Identificador único
- `codigo`: Código único (ex: PROJ-2024-001) para rastreamento
- `gerente_id`: FK para FUNCIONÁRIO (quem gerencia o projeto)
- `status`: Controla fase do projeto (Planejamento → Em Andamento → Concluído)
- `orcamento`: Para controle financeiro

**Por que gerente é um FUNCIONÁRIO?**
- Gerente é sempre um funcionário
- Permite rastrear quem lidera cada projeto
- Cria relacionamento 1:N: Um funcionário pode gerenciar vários projetos

---

#### **Tabela ALOCACAO (Associativa - IMPORTANTE!)**

**Propósito:** Conectar funcionários com projetos e armazenar dados da participação

**Campos críticos:**

| Campo | Propósito |
|-------|----------|
| `id_alocacao` | ID único de cada alocação |
| `id_funcionario` | Qual funcionário |
| `id_projeto` | Em qual projeto |
| `data_inicio` | **OBRIGATÓRIO (requisito do exercício)** |
| `data_fim` | NULL = ainda alocado; preenchido = saiu |
| `funcao_projeto` | Qual é o papel (Desenvolvedor, Tester, Arquiteto) |
| `horas_alocadas_semana` | Dedicação (20h, 40h, etc.) |
| `status` | Alocado, Pausa, Finalizado |
| `salario_projeto` | Taxa específica neste projeto |

**Por que UNIQUE em (id_funcionario, id_projeto, data_inicio)?**
- Evita duplicação: mesmo funcionário não entra 2x no mesmo projeto na mesma data
- Permite histórico: se funcionário sair e voltar, cria novo registro com data_inicio diferente

## 7. Exemplos de Consultas SQL

````python
-- CONSULTA 1: Listar todos os projetos de um funcionário específico
SELECT 
    f.nome AS funcionario,
    p.codigo,
    p.nome AS projeto,
    a.data_inicio,
    a.data_fim,
    a.funcao_projeto,
    a.horas_alocadas_semana,
    a.status AS status_alocacao
FROM funcionario f
INNER JOIN alocacao a ON f.id_funcionario = a.id_funcionario
INNER JOIN projeto p ON a.id_projeto = p.id_projeto
WHERE f.id_funcionario = 2
ORDER BY a.data_inicio DESC;

-- CONSULTA 2: Listar todos os funcionários de um projeto específico
SELECT 
    f.nome,
    f.cargo,
    a.funcao_projeto,
    a.data_inicio,
    a.data_fim,
    a.horas_alocadas_semana,
    a.status
FROM projeto p
INNER JOIN alocacao a ON p.id_projeto = a.id_projeto
INNER JOIN funcionario f ON a.id_funcionario = f.id_funcionario
WHERE p.id_projeto = 1
ORDER BY a.data_inicio;

-- CONSULTA 3: Contar quantos projetos cada funcionário participa
SELECT 
    f.nome,
    f.cargo,
    COUNT(DISTINCT a.id_projeto) AS total_projetos,
    SUM(a.horas_alocadas_semana) AS total_horas_semana
FROM funcionario f
LEFT JOIN alocacao a ON f.id_funcionario = a.id_funcionario AND a.status = 'Alocado'
GROUP BY f.id_funcionario, f.nome, f.cargo
ORDER BY total_projetos DESC;

-- CONSULTA 4: Listar projetos e quantos funcionários têm alocação ativa
SELECT 
    p.codigo,
    p.nome AS projeto,
    p.status,
    COUNT(DISTINCT a.id_funcionario) AS total_funcionarios,
    GROUP_CONCAT(DISTINCT f.nome) AS nomes_funcionarios
FROM projeto p
LEFT JOIN alocacao a ON p.id_projeto = a.id_projeto AND a.status = 'Alocado'
LEFT JOIN funcionario f ON a.id_funcionario = f.id_funcionario
GROUP BY p.id_projeto, p.codigo, p.nome, p.status
ORDER BY p.data_inicio;

-- CONSULTA 5: Funcionários que saíram de um projeto (data_fim preenchida)
SELECT 
    f.nome,
    p.codigo,
    p.nome AS projeto,
    a.data_inicio,
    a.data_fim,
    a.funcao_projeto,
    DATEDIFF(a.data_fim, a.data_inicio) AS dias_trabalhados
FROM funcionario f
INNER JOIN alocacao a ON f.id_funcionario = a.id_funcionario
INNER JOIN projeto p ON a.id_projeto = p.id_projeto
WHERE a.data_fim IS NOT NULL
ORDER BY a.data_fim DESC;

-- CONSULTA 6: Custo total de alocações por projeto
SELECT 
    p.codigo,
    p.nome AS projeto,
    p.orcamento,
    SUM(a.salario_projeto) AS salario_total_alocado,
    COUNT(DISTINCT a.id_funcionario) AS qtd_funcionarios,
    (p.orcamento - SUM(a.salario_projeto)) AS margem_disponivel
FROM projeto p
LEFT JOIN alocacao a ON p.id_projeto = a.id_projeto
GROUP BY p.id_projeto, p.codigo, p.nome, p.orcamento
ORDER BY margem_disponivel;
````

# Exercício 4 - Modelo de Dados: Sistema de Consultas em Clínica

## Problema
Em uma clínica, cada consulta é feita por exatamente um médico, mas um médico pode realizar várias consultas. Cada consulta está vinculada a um único paciente, mas o paciente pode ter várias consultas ao longo do tempo.

**Crie o diagrama conceitual com cardinalidades mínimas e máximas**

## 1. Identificação das Entidades

### **Entidade 1: MÉDICO**
- Representa um profissional da saúde que realiza consultas

### **Entidade 2: PACIENTE**
- Representa uma pessoa que busca atendimento na clínica

### **Entidade 3: CONSULTA**
- Representa o encontro entre médico e paciente em um horário específico

## 2. Análise das Cardinalidades

### **Relacionamento 1: MÉDICO — CONSULTA**

**Descrição:** "Um médico pode realizar várias consultas, mas cada consulta é feita por exatamente um médico"

```
MÉDICO (1) ────────────────────── (0..N) CONSULTA
   │
   ├─ Mínimo: 0  → Médico pode não ter consultas (novo, afastado, etc.)
   └─ Máximo: N  → Médico pode ter muitas consultas
   
CONSULTA (N) ────────────────────── (1) MÉDICO
   │
   ├─ Mínimo: 1  → Consulta OBRIGATORIAMENTE tem um médico
   └─ Máximo: 1  → Consulta tem EXATAMENTE um médico (não compartilhada)
```

**Tipo:** **1:N (Um para Muitos)**
- Funcional: CONSULTA.id_medico referencia MÉDICO.id_medico (Foreign Key)

---

### **Relacionamento 2: PACIENTE — CONSULTA**

**Descrição:** "Um paciente pode ter várias consultas ao longo do tempo, mas cada consulta está vinculada a um único paciente"

```
PACIENTE (1) ────────────────────── (0..N) CONSULTA
   │
   ├─ Mínimo: 0  → Paciente pode não ter consultas agendadas
   └─ Máximo: N  → Paciente pode ter múltiplas consultas

CONSULTA (N) ────────────────────── (1) PACIENTE
   │
   ├─ Mínimo: 1  → Consulta OBRIGATORIAMENTE tem um paciente
   └─ Máximo: 1  → Consulta atende EXATAMENTE um paciente
```

**Tipo:** **1:N (Um para Muitos)**
- Funcional: CONSULTA.id_paciente referencia PACIENTE.id_paciente (Foreign Key)

---

### **Resumo das Cardinalidades:**

| Relacionamento | Tipo | Min-Max Médico | Min-Max Paciente | Min-Max Consulta |
|---|---|---|---|---|
| MÉDICO ← → CONSULTA | 1:N | (0,N) | N/A | (1,1) |
| PACIENTE ← → CONSULTA | 1:N | N/A | (0,N) | (1,1) |

**Interpretação:**
- ✅ Cada consulta obriga ter 1 médico E 1 paciente
- ✅ Um médico pode ter 0 ou muitas consultas
- ✅ Um paciente pode ter 0 ou muitas consultas
- ✅ Uma consulta NÃO pode existir sem médico ou paciente

## 3. Diagrama Conceitual com Cardinalidades

### **Notação de Cardinalidade (Min, Max)**

```
(0,1)  ──  uma entidade tem no mínimo 0 e no máximo 1
(1,1)  ──  uma entidade tem no mínimo 1 e no máximo 1 (obrigatório e único)
(0,N)  ──  uma entidade tem no mínimo 0 e no máximo vários
(1,N)  ──  uma entidade tem no mínimo 1 e no máximo vários
```

---

### **Diagrama Entidade-Relacionamento Conceitual**

```
┌─────────────────────┐
│      MÉDICO         │
├─────────────────────┤
│ • id_medico (PK)    │
│ • nome              │

          │
```

### **Leitura do Diagrama:**

**1. Relacionamento MÉDICO → CONSULTA**
```
Um MÉDICO pode realizar (0,N) CONSULTAS
   └─ 0 = Médico novo, afastado ou sem consultas agendadas
   └─ N = Médico experiente com muitas consultas
```

**2. Relacionamento PACIENTE → CONSULTA**
```
Um PACIENTE pode ter (0,N) CONSULTAS
   └─ 0 = Paciente novo, sem consultas agendadas ainda
   └─ N = Paciente com histórico de múltiplas visitas
```

**3. Relacionamento CONSULTA → MÉDICO**
```
Uma CONSULTA é realizada por (1,1) MÉDICO
   └─ 1 = Mínimo 1 médico (obrigatório)
   └─ 1 = Máximo 1 médico (consulta não compartilhada)
```

**4. Relacionamento CONSULTA → PACIENTE**
```
Uma CONSULTA atende (1,1) PACIENTE
   └─ 1 = Mínimo 1 paciente (obrigatório)
   └─ 1 = Máximo 1 paciente (consulta individual)
```

### **Por que essas cardinalidades?**

|---------|---------------|
| Médico (0,N) | Médico pode não ter consultas (novo, férias, etc.) |
| Paciente (0,N) | Paciente pode nunca ter consultado (cadastro apenas) |
| Consulta com 1 Médico | Cada consulta necessita exatamente um profissional |
| Consulta com 1 Paciente | Consulta individual, não em grupo |
| Consulta mínimo 1 de tudo | Uma consulta não pode existir sem referência |

## 5. Código SQL - Criação das Tabelas

````python
-- Tabela MÉDICO
CREATE TABLE medico (
    id_medico INT PRIMARY KEY AUTO_INCREMENT,
    nome VARCHAR(100) NOT NULL,
    crm VARCHAR(20) NOT NULL UNIQUE,
    especialidade VARCHAR(50) NOT NULL,
    data_nascimento DATE NOT NULL,
    email VARCHAR(100) UNIQUE,
    telefone VARCHAR(15),
    status VARCHAR(20) DEFAULT 'Ativo',
    data_cadastro TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);

-- Tabela PACIENTE
CREATE TABLE paciente (
    id_paciente INT PRIMARY KEY AUTO_INCREMENT,
    nome VARCHAR(100) NOT NULL,
    cpf VARCHAR(11) NOT NULL UNIQUE,
    data_nascimento DATE NOT NULL,
    email VARCHAR(100),
    telefone VARCHAR(15) NOT NULL,
    endereco VARCHAR(200),
    data_cadastro TIMESTAMP DEFAULT CURRENT_TIMESTAMP
);

-- Tabela CONSULTA
CREATE TABLE consulta (
    id_consulta INT PRIMARY KEY AUTO_INCREMENT,
    id_medico INT NOT NULL,
    id_paciente INT NOT NULL,
    data_hora DATETIME NOT NULL UNIQUE,
    duracao_minutos INT DEFAULT 30,
    diagnostico TEXT,
    prescricao TEXT,
    status VARCHAR(20) DEFAULT 'Agendada',
    valor_consulta DECIMAL(10,2),
    data_criacao TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (id_medico) REFERENCES medico(id_medico),
    FOREIGN KEY (id_paciente) REFERENCES paciente(id_paciente),
    INDEX idx_medico (id_medico),
    INDEX idx_paciente (id_paciente),
    INDEX idx_data_hora (data_hora)
);
````

### **Análise das Tabelas**

#### **Tabela MÉDICO**
- **Chave Primária:** `id_medico`
- **Chaves Únicas:** `crm` (número de registro no conselho)
- **Relacionamento:** Uma tabela MÉDICO pode ter 0 a N registros em CONSULTA
- **Observações:** 
  - CRM é obrigatório e único (identidade profissional)
  - Status controla se médico está disponível

#### **Tabela PACIENTE**
- **Chave Primária:** `id_paciente`
- **Chaves Únicas:** `cpf` (documento de identidade)
- **Relacionamento:** Uma tabela PACIENTE pode ter 0 a N registros em CONSULTA
- **Observações:**
  - CPF é obrigatório e único
  - Telefone é obrigatório para agendamentos

#### **Tabela CONSULTA**
- **Chave Primária:** `id_consulta`
- **Chaves Estrangeiras:**
  - `id_medico`: Referencia MÉDICO (obrigatório - NOT NULL)
  - `id_paciente`: Referencia PACIENTE (obrigatório - NOT NULL)
- **Chave Única:** `data_hora` (não pode haver 2 consultas no mesmo horário)
- **Índices:** Criados para melhorar performance de buscas
- **Observações:**
  - Cada consulta **OBRIGA** ter médico e paciente
  - Sem médico ou paciente = erro no banco (integridade referencial)
  - Status rastreia: Agendada → Realizada → Cancelada

# Exercício 5 - Modelo de Dados: Sistema de E-commerce

## Problema
Um sistema de e-commerce precisa registrar clientes, pedidos e produtos com os seguintes requisitos:
- O cliente faz vários pedidos
- Cada pedido pode conter vários produtos
- Cada produto pode aparecer em vários pedidos
- Para cada produto em um pedido, é necessário guardar a quantidade e o valor unitário

**Objetivo:** Montar o DER (Diagrama de Entidade-Relacionamento) completo.

## 1. Identificação das Entidades

### **Entidade 1: CLIENTE**
- Representa uma pessoa que realiza compras no e-commerce
- Atributos: id_cliente (PK), nome, email, telefone, endereco, data_cadastro

### **Entidade 2: PEDIDO**
- Representa uma compra realizada por um cliente em um determinado momento
- Atributos: id_pedido (PK), id_cliente (FK), data_pedido, status, valor_total

### **Entidade 3: PRODUTO**
- Representa um item que pode ser comprado no e-commerce
- Atributos: id_produto (PK), nome, descricao, preco_unitario, estoque

### **Entidade 4: ITEM_PEDIDO** (Tabela de Junção/Associação)
- Representa a associação entre PEDIDO e PRODUTO
- Armazena a quantidade e valor unitário de cada produto em um pedido
- Atributos: id_item (PK), id_pedido (FK), id_produto (FK), quantidade, valor_unitario

## 2. Diagrama de Entidade-Relacionamento (DER)

## 3. Análise Detalhada das Cardinalidades e Relacionamentos

### **Relacionamento 1: CLIENTE — PEDIDO (REALIZA)**

**Descrição:** "Um cliente pode fazer vários pedidos, mas cada pedido pertence a exatamente um cliente"

```
CLIENTE (1) ───────────────────────────── (0..N) PEDIDO
    │
    ├─ Mínimo: 0  → Um cliente pode estar cadastrado sem ter realizado pedidos
    └─ Máximo: N  → Um cliente pode fazer múltiplos pedidos ao longo do tempo

PEDIDO (N) ───────────────────────────── (1) CLIENTE
    │
    ├─ Mínimo: 1  → Um pedido OBRIGATORIAMENTE pertence a um cliente
    └─ Máximo: 1  → Um pedido pertence a EXATAMENTE um cliente (não compartilhado)
```

**Tipo:** **1:N (Um para Muitos)**
- **Funcional:** PEDIDO.id_cliente referencia CLIENTE.id_cliente (Foreign Key)
- **Implementação:** Foreign Key não nula em PEDIDO

---

### **Relacionamento 2: PEDIDO — ITEM_PEDIDO (CONTÉM)**

**Descrição:** "Um pedido pode conter vários itens (produtos), mas cada item pertence a exatamente um pedido"

```
PEDIDO (1) ───────────────────────────── (1..N) ITEM_PEDIDO
    │
    ├─ Mínimo: 1  → Um pedido OBRIGATORIAMENTE contém pelo menos 1 item
    └─ Máximo: N  → Um pedido pode ter múltiplos itens/produtos

ITEM_PEDIDO (N) ───────────────────────────── (1) PEDIDO
    │
    ├─ Mínimo: 1  → Um item OBRIGATORIAMENTE pertence a um pedido
    └─ Máximo: 1  → Um item pertence a EXATAMENTE um pedido
```

**Tipo:** **1:N (Um para Muitos)**
- **Funcional:** ITEM_PEDIDO.id_pedido referencia PEDIDO.id_pedido (Foreign Key)
- **Implementação:** Foreign Key não nula em ITEM_PEDIDO

---

### **Relacionamento 3: PRODUTO — ITEM_PEDIDO (CONTÉM)**

**Descrição:** "Um produto pode aparecer em vários itens de pedidos diferentes, mas cada item referencia exatamente um produto"

```
PRODUTO (1) ───────────────────────────── (0..N) ITEM_PEDIDO
    │
    ├─ Mínimo: 0  → Um produto pode existir no catálogo sem ter sido pedido
    └─ Máximo: N  → Um produto pode ser solicitado em múltiplos pedidos

ITEM_PEDIDO (N) ───────────────────────────── (1) PRODUTO
    │
    ├─ Mínimo: 1  → Um item OBRIGATORIAMENTE referencia um produto
    └─ Máximo: 1  → Um item referencia EXATAMENTE um produto
```

**Tipo:** **1:N (Um para Muitos)**
- **Funcional:** ITEM_PEDIDO.id_produto referencia PRODUTO.id_produto (Foreign Key)
- **Implementação:** Foreign Key não nula em ITEM_PEDIDO

---

### **Relacionamento IMPLÍCITO: CLIENTE — PRODUTO (M:N através de PEDIDO)**

Existe um relacionamento **N:M (Muitos para Muitos)** indireto entre CLIENTE e PRODUTO:
- Um cliente pode comprar vários produtos
- Um produto pode ser comprado por vários clientes
- Esse relacionamento é materializado através da tabela ITEM_PEDIDO

```
CLIENTE (N) ──────────[via PEDIDO]──────── (N) PRODUTO

Caminho: CLIENTE → PEDIDO → ITEM_PEDIDO → PRODUTO
```

---

### **Resumo das Cardinalidades**

| Relacionamento | Tipo | Descrição | Cardinalidade |
|---|---|---|---|
| CLIENTE — PEDIDO | 1:N | Cliente realiza pedidos | (0,N) : (1,1) |
| PEDIDO — ITEM_PEDIDO | 1:N | Pedido contém itens | (1,N) : (1,1) |
| PRODUTO — ITEM_PEDIDO | 1:N | Produto aparece em itens | (0,N) : (1,1) |
| CLIENTE ⟶ PRODUTO (indireto) | N:M | Via tabela de junção | (0,N) : (0,N) |

**Observações Importantes:**
- ✅ ITEM_PEDIDO é uma **tabela de junção/associação** que resolve o relacionamento N:M
- ✅ Armazena informações da associação: quantidade e valor_unitario
- ✅ Um pedido não pode estar vazio (mínimo 1 item)
- ✅ Um cliente pode não ter pedidos (novo cliente)
- ✅ Um produto pode não ter sido pedido (produto novo)

## 4. Modelo Lógico - Transformação para Tabelas SQL

### **Tabela: CLIENTE**
```sql
CREATE TABLE cliente (
    id_cliente INT PRIMARY KEY AUTO_INCREMENT,
    nome VARCHAR(100) NOT NULL,
    email VARCHAR(100) UNIQUE NOT NULL,
    telefone VARCHAR(20),
    endereco VARCHAR(200) NOT NULL,
    data_cadastro DATETIME DEFAULT CURRENT_TIMESTAMP
);
```

### **Tabela: PEDIDO**
```sql
CREATE TABLE pedido (
    id_pedido INT PRIMARY KEY AUTO_INCREMENT,
    id_cliente INT NOT NULL,
    data_pedido DATETIME DEFAULT CURRENT_TIMESTAMP,
    status VARCHAR(20) DEFAULT 'Pendente',  -- Pendente, Processando, Enviado, Entregue, Cancelado
    valor_total DECIMAL(10, 2) DEFAULT 0,
    FOREIGN KEY (id_cliente) REFERENCES cliente(id_cliente)
);
```

### **Tabela: PRODUTO**
```sql
CREATE TABLE produto (
    id_produto INT PRIMARY KEY AUTO_INCREMENT,
    nome VARCHAR(100) NOT NULL,
    descricao TEXT,
    preco_unitario DECIMAL(10, 2) NOT NULL,
    estoque INT DEFAULT 0
);
```

### **Tabela: ITEM_PEDIDO** (Tabela de Junção/Associação)
```sql
CREATE TABLE item_pedido (
    id_item INT PRIMARY KEY AUTO_INCREMENT,
    id_pedido INT NOT NULL,
    id_produto INT NOT NULL,
    quantidade INT NOT NULL CHECK (quantidade > 0),
    valor_unitario DECIMAL(10, 2) NOT NULL,  -- Preço no momento da compra
    FOREIGN KEY (id_pedido) REFERENCES pedido(id_pedido) ON DELETE CASCADE,
    FOREIGN KEY (id_produto) REFERENCES produto(id_produto),
    UNIQUE KEY (id_pedido, id_produto)  -- Evita duplicar o mesmo produto no pedido
);
```

### **Chaves e Relacionamentos**

| Tabela | Tipo de Chave | Descrição |
|---|---|---|
| CLIENTE | PK: id_cliente | Identifica unicamente cada cliente |
| PEDIDO | PK: id_pedido | Identifica unicamente cada pedido |
| PEDIDO | FK: id_cliente | Referencia o cliente que fez o pedido |
| PRODUTO | PK: id_produto | Identifica unicamente cada produto |
| ITEM_PEDIDO | PK: id_item | Identifica unicamente cada item de pedido |
| ITEM_PEDIDO | FK: id_pedido | Referencia o pedido ao qual o item pertence |
| ITEM_PEDIDO | FK: id_produto | Referencia o produto do item |
| ITEM_PEDIDO | UNQ: (id_pedido, id_produto) | Garante que não há duplicação de produtos no mesmo pedido |

```sql
UPDATE pedido p
SET valor_total = (
    SELECT SUM(ip.quantidade * ip.valor_unitario)
    FROM item_pedido ip
    WHERE ip.id_pedido = p.id_pedido
)
WHERE p.id_pedido = 1;
```

## 7. Resumo e Pontos-Chave

### **Estrutura do DER - Componentes Principais**

1. **Entidades Independentes:**
   - CLIENTE: Armazena informações dos compradores
   - PRODUTO: Catálogo de itens disponíveis para venda

2. **Entidade Dependente:**
   - PEDIDO: Vinculada a CLIENTE (relacionamento 1:N)
   
3. **Tabela de Junção/Associação:**
   - ITEM_PEDIDO: Resolve o relacionamento N:M entre PEDIDO e PRODUTO

### **Relacionamentos e Cardinalidades**

| PEDIDO → ITEM_PEDIDO | 1:N | (1,N):(1,1) | Pedido deve ter pelo menos 1 item |
| PRODUTO → ITEM_PEDIDO | 1:N | (0,N):(1,1) | Produto pode nunca ter sido vendido |
| CLIENTE ⇄ PRODUTO (indireto) | N:M | (0,N):(0,N) | Através de PEDIDO e ITEM_PEDIDO |

### **Razão da Tabela ITEM_PEDIDO**

A tabela ITEM_PEDIDO é **essencial** porque:

1. **Resolve Relacionamento N:M:** Um pedido pode ter vários produtos, um produto pode estar em vários pedidos
2. **Armazena Informações de Contexto:** 
   - `quantidade`: Quantos produtos foram pedidos

3. **Mantém Histórico:** Se o preço do produto mudar, o histórico de vendas anteriores permanece intacto

# Exercício 6 - Modelo de Dados: Sistema de Cadastro de Veículos

## Problema
Uma instituição precisa cadastrar veículos com os seguintes requisitos:
- Existem dois tipos de veículos: **CARRO** e **MOTOCICLETA**
- Todo veículo tem: **placa** e **ano**
- Carro tem: **número de portas**
- Motocicleta tem: **cilindrada**

**Objetivo:** Modelar esse cenário com especialização/herança de entidades.

## 1. Identificação das Entidades

### **Entidade Principal: VEÍCULO**
- Representa a classe genérica que engloba todos os veículos
- Atributos **comuns** a todos os veículos:
  - id_veiculo (PK) - Identificador único
  - placa - Identificação do veículo
  - ano - Ano de fabricação
  - tipo_veiculo - Discriminador (CARRO ou MOTOCICLETA)

### **Entidade Especializada 1: CARRO** (Herança de VEÍCULO)
- Representa veículos de 4 rodas
- Herda atributos de VEÍCULO
- Atributo específico:
  - num_portas - Quantidade de portas (2, 4, etc.)

### **Entidade Especializada 2: MOTOCICLETA** (Herança de VEÍCULO)
- Representa veículos de 2 rodas
- Herda atributos de VEÍCULO
- Atributo específico:
  - cilindrada - Potência do motor em cc (500, 1000, 1200, etc.)

## 3. Análise da Especialização

### **Conceito: Especialização/Herança**

A especialização (também chamada de herança ou generalização) é um relacionamento entre uma entidade genérica (superentidade) e entidades mais específicas (subentidades).

| Aspecto | Descrição |
|---|---|
| **Superentidade** | VEÍCULO - Contém atributos e características comuns |
| **Subentidades** | CARRO, MOTOCICLETA - Herdam de VEÍCULO e adicionam atributos específicos |
| **Atributos Herdados** | placa, ano (de VEÍCULO) |
| **Atributos Específicos** | num_portas (CARRO), cilindrada (MOTOCICLETA) |

### **Tipo de Especialização: TOTAL e DISJUNTA**

```
TOTAL (ou Completa):
  └─ TODO veículo é CARRO OU MOTOCICLETA (não existe veículo genérico)
  └─ Não há exceções

DISJUNTA (ou Exclusiva):
  └─ Um veículo é CARRO OU MOTOCICLETA (não é ambos simultaneamente)
  └─ Não há sobreposição entre os subtipos
```

### **Cardinalidade da Especialização**

```
VEÍCULO (1) ──────────[ESPECIALIZA]──────── (1) CARRO
   │
   ├─ Total: Cada CARRO é exatamente 1 VEÍCULO
   └─ Disjunta: Um CARRO não pode ser MOTOCICLETA

VEÍCULO (1) ──────────[ESPECIALIZA]──────── (1) MOTOCICLETA
   │
   ├─ Total: Cada MOTOCICLETA é exatamente 1 VEÍCULO
   └─ Disjunta: Uma MOTOCICLETA não pode ser CARRO
```

### **Interpretação das Cardinalidades**

| Relacionamento | Min | Max | Significado |
|---|---|---|---|
| CARRO → VEÍCULO | 1 | 1 | Todo CARRO referencia exatamente 1 VEÍCULO |
| MOTOCICLETA → VEÍCULO | 1 | 1 | Toda MOTOCICLETA referencia exatamente 1 VEÍCULO |
| VEÍCULO → CARRO | 0 | 1 | Um VEÍCULO pode ser 0 ou 1 CARRO (nunca mais de 1) |
| VEÍCULO → MOTOCICLETA | 0 | 1 | Um VEÍCULO pode ser 0 ou 1 MOTOCICLETA (nunca mais de 1) |

## 4. Implementação SQL - 3 Abordagens Diferentes

### **ABORDAGEM 1: Tabelas Separadas com Herança Explícita**

**Modelo:** Uma tabela para a superentidade + uma tabela para cada subtipo

```sql
-- Tabela Superentidade (VEÍCULO)
CREATE TABLE veiculo (
    id_veiculo INT PRIMARY KEY AUTO_INCREMENT,
    placa VARCHAR(10) UNIQUE NOT NULL,
    ano INT NOT NULL,
    tipo_veiculo ENUM('CARRO', 'MOTOCICLETA') NOT NULL

);
```

## 7. Resumo e Conceitos-Chave

### **O que é Especialização/Herança em BD?**

Especialização é um relacionamento entre uma entidade genérica (superentidade) e entidades mais específicas (subentidades), onde:

1. **Herança de Atributos:** Subentidades herdam todos os atributos da superentidade
2. **Atributos Próprios:** Subentidades podem ter atributos exclusivos
3. **Restrições:** Existem restrições sobre quantas subentidades um registro pode pertencer

---

### **Tipos de Especialização**

| Tipo | Descrição | Exemplo |
|---|---|---|
| **Total** | Todo registro da superentidade pertence a alguma subentidade | Todo VEÍCULO é CARRO ou MOTOCICLETA |
| **Parcial** | Nem todo registro precisa pertencer a uma subentidade | Nem todo PESSOA precisa ser FUNCIONÁRIO |
| **Disjunta** | Um registro pertence a NO MÁXIMO UMA subentidade | Um VEÍCULO é OU CARRO OU MOTOCICLETA |
| **Sobreposta** | Um registro pode pertencer a MÚLTIPLAS subentidades | Uma PESSOA pode ser PROFESSOR E PESQUISADOR |

---

### **Vantagens da Especialização**

✅ Melhor organização e estrutura de dados  
✅ Facilita restrições específicas por tipo  
✅ Evita dados NULL desnecessários  
✅ Normalização mais correta  
✅ Consultas e manutenção mais claras  

---

### **Comparação das Abordagens**

| Abordagem | Normalização | Simplicidade | Flexibilidade | Espaço | Recomendação |
|---|---|---|---|---|---|
| **1. Tabelas Separadas** | ⭐⭐⭐⭐⭐ | ⭐⭐⭐ | ⭐⭐ | ⭐⭐⭐⭐ | ✅ Melhor |
| **2. Tabela Única** | ⭐⭐ | ⭐⭐⭐⭐⭐ | ⭐⭐⭐ | ⭐⭐ | ⚠️ Rápido prototipagem |
| **3. JSON** | ⭐ | ⭐⭐⭐⭐ | ⭐⭐⭐⭐⭐ | ⭐⭐⭐ | ⭐ Sistemas modernos |

---

### **Checklist: Quando Usar Especialização?**

✅ Existem múltiplos "tipos" de uma entidade  
✅ Tipos têm atributos distintos  
✅ Tipos têm comportamentos/restrições diferentes  
✅ Relações com outras entidades diferem por tipo  
✅ Necessidade de consultas específicas por tipo

- **PESSOA:** ├─ CLIENTE, ├─ FORNECEDOR, └─ FUNCIONÁRIO

### **Observações Importantes**

**Sobre o Discriminador (tipo_veiculo):**
- Armazena qual é o subtipo
- Usado para determinar qual tabela de subtipo consultar
- Implementado como ENUM para garantir valores válidos
- Permite auditoria e validação

**Sobre as Foreign Keys:**
- ON DELETE CASCADE garante que ao deletar VEÍCULO, suas entradas em CARRO/MOTOCICLETA também sejam deletadas
- PRIMARY KEY da subentidade é também FOREIGN KEY que referencia a superentidade

**Sobre os Triggers:**
- Validam que um carro está inserido apenas como CARRO, não como MOTOCICLETA
- Garantem integridade referencial mais forte
- Evitam dados inconsistentes
